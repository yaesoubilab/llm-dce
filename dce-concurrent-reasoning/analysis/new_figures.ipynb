## Helper Functions

```python
def extract_strategy_number(strategy_name):
    """Extract numeric value from strategy name for proper sorting."""
    word_to_num = {
        'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5,
        'six': 6, 'seven': 7, 'eight': 8, 'nine': 9, 'ten': 10
    }
    strategy_lower = str(strategy_name).lower()
    digits = ''.join(filter(str.isdigit, strategy_lower))
    if digits:
        return int(digits)
    for word, num in word_to_num.items():
        if word in strategy_lower:
            return num
    return 0

def format_strategy_name(strategy_name):
    """Format strategy name to use words instead of digits."""
    num_to_word = {
        1: "One", 2: "Two", 3: "Three", 4: "Four", 5: "Five",
        6: "Six", 7: "Seven", 8: "Eight", 9: "Nine", 10: "Ten"
    }
    strategy_num = extract_strategy_number(strategy_name)
    if strategy_num in num_to_word:
        return f"Strategy {num_to_word[strategy_num]}"
    return str(strategy_name).replace("_", " ").title()

def add_panel_labels(axes, labels=None, x_offset=-0.02, y_offset=1.05):
    """Add letter labels (A, B, C...) and optional subtitles to subplots."""
    axes_flat = axes.flatten() if isinstance(axes, np.ndarray) else [axes]
    for i, ax in enumerate(axes_flat):
        if ax.get_visible():
            char = string.ascii_uppercase[i]
            text = char
            if labels and i < len(labels):
                text = f"{char}) {labels[i]}"
            
            ax.text(x_offset, y_offset, text, transform=ax.transAxes,
                    fontsize=26, fontweight='bold', va='bottom', 
                    ha='right' if not labels else 'left')

def get_context_column(strategy_name):
    """Get context column for a given strategy."""
    strategy_num = extract_strategy_number(strategy_name)
    context_mapping = {
        2: 'capacity_context',
        3: 'income_loss_selected',
        4: 'user_preference_type'
    }
    return context_mapping.get(strategy_num, None)

def get_legend_title(context_col):
    """Get human-readable legend title for context column."""
    if not context_col: return None
    title_mapping = {
        'capacity_context': 'Capacity Context',
        'income_loss_selected': 'Income Loss',
        'user_preference_type': 'User Preference'
    }
    return title_mapping.get(context_col, context_col.replace('_', ' ').title())

def get_column_label(col):
    """Get human-readable label for reasoning component."""
    mapping = REASONING_COMPONENT_CONFIG
    return mapping.get(col, col.replace('_', ' ').title())

def calculate_errorbar_data(df, response_col, context_cols, use_percentage=False):
    """Calculate means and 95% confidence intervals."""
    grouped = df.groupby(context_cols)[response_col]
    plot_data = []
    for name, group in grouped:
        if len(group) > 0:
            data = group.values
            if use_percentage: 
                data = data * 100
            mean_val = data.mean()
            if len(data) > 1:
                ci = stats.t.interval(0.95, len(data) - 1, loc=mean_val, scale=stats.sem(data))
                ci_lower, ci_upper = mean_val - ci[0], ci[1] - mean_val
            else:
                ci_lower, ci_upper = 0, 0
                
            record = {}
            if isinstance(name, tuple):
                for i, col in enumerate(context_cols): 
                    record[col] = name[i]
            else:
                record[context_cols[0]] = name
            record['mean'] = mean_val
            record['ci_lower'] = ci_lower
            record['ci_upper'] = ci_upper
            plot_data.append(record)
    return pd.DataFrame(plot_data)

print("Helper functions loaded")
```

## Function

```python
def run_rf_all_strategies_comparison(
    results_df,
    min_obs=10,
    show_tables=True,
    plot=True,
    # Random Forest params
    n_estimators=500,
    max_depth=None,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1,
    # CV params
    cv_folds=10
):
    """
    Random Forest classifier for all 4 strategies with strategy-specific features.
    
    Strategy 1: inf_diff_z, bc_diff_z
    Strategy 2: inf_diff_z, bc_diff_z, cap_z
    Strategy 3: inf_diff_z, bc_diff_z, cap_z, income_loss_diff_z
    Strategy 4: inf_diff_z, bc_diff_z, cap_z, income_loss_diff_z, preferred_option_num_z
    
    Outputs:
      - all_results: dict with per_model_df, overall_df, importances_df for each strategy
      - Visualization: 2x2 grid of feature importance bar charts (one per strategy)
    """
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.metrics import roc_auc_score
    from sklearn.model_selection import StratifiedKFold
    
    
    # --- Styling Configuration ---
    MODEL_NAME_MAPPING = {
        'x-ai/grok-4-fast': 'Grok',
        'openai/gpt-5-nano': 'ChatGPT',
        'google/gemini-2.5-flash': 'Gemini',
        'deepseek/deepseek-chat-v3.1': 'DeepSeek'
    }
    
    # Fixed alphabetical order for models
    MODEL_ORDER = ['ChatGPT', 'DeepSeek', 'Gemini', 'Grok']
    
    _palette_raw = sns.color_palette("tab10", n_colors=len(MODEL_ORDER))
    MODEL_COLORS = dict(zip(MODEL_ORDER, _palette_raw))
    
    FIG_SIZE_MULTI = (18, 14)
    FONT_SCALE = 2.0
    DPI = 600
    
    sns.set_style("white")
    sns.set_context("paper", font_scale=FONT_SCALE)
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']
    plt.rcParams['figure.dpi'] = DPI
    
    
    # --- Helper: NumPy z-score ---
    def np_zscore(series: pd.Series) -> pd.Series:
        arr = np.asarray(series, dtype=float)
        mu = np.nanmean(arr)
        sigma = np.nanstd(arr, ddof=0)
        if (not np.isfinite(sigma)) or sigma == 0:
            return pd.Series(np.full_like(arr, np.nan, dtype=float), index=series.index)
        return pd.Series((arr - mu) / sigma, index=series.index)
    
    
    # --- Display helper ---
    def show_df(df, title):
        if not show_tables:
            return
        print("\n" + "=" * 60)
        print(title)
        print("=" * 60)
        try:
            from IPython.display import display
            display(df)
        except Exception:
            print(df.to_string(index=False))
    
    
    # --- Strategy-specific feature definitions (in order) ---
    strategy_features = {
        "strategy_one": ["inf_diff_z", "bc_diff_z"],
        "strategy_two": ["inf_diff_z", "bc_diff_z", "cap_z"],
        "strategy_three": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z"],
        "strategy_four": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z", "preferred_option_num_z"]
    }
    
    strategy_titles = {
        "strategy_one": "Strategy 1",
        "strategy_two": "Strategy 2",
        "strategy_three": "Strategy 3",
        "strategy_four": "Strategy 4"
    }
    
    
    def add_features(m_df: pd.DataFrame, features: list) -> pd.DataFrame:
        m_df = m_df.copy()
        
        # Z-score all potential features
        m_df["inf_diff_z"] = np_zscore(m_df["inf_diff"])
        m_df["bc_diff_z"]  = np_zscore(m_df["bc_diff"])
        
        if "cap_z" in features:
            m_df["cap_z"] = np_zscore(m_df["capacity_context"])
        
        if "income_loss_diff_z" in features:
            # Exogenous income loss differential: option_1 minus option_2, independent of choice
            m_df["income_loss_1"] = np.where(m_df["choice"] == 1, m_df["income_loss_selected"], m_df["income_loss_non_selected"])
            m_df["income_loss_2"] = np.where(m_df["choice"] == 1, m_df["income_loss_non_selected"], m_df["income_loss_selected"])
            m_df["income_loss_diff"] = m_df["income_loss_1"] - m_df["income_loss_2"]
            m_df["income_loss_diff_z"] = np_zscore(m_df["income_loss_diff"])
        
        if "preferred_option_num_z" in features:
            m_df["preferred_option_num_z"] = np_zscore(m_df["preferred_option_num"])
        
        return m_df.dropna(subset=["choice"] + features)
    
    
    # --- Process each strategy ---
    all_results = {}
    
    for strategy_name, features in strategy_features.items():
        print(f"\n{'='*60}")
        print(f"Processing {strategy_name}")
        print(f"{'='*60}")
        
        # Filter strategy
        df = results_df[results_df["strategy"] == strategy_name].copy()
        print(f"Found {len(df)} rows for {strategy_name}")
        
        if df.empty:
            all_results[strategy_name] = {
                "per_model_df": pd.DataFrame(),
                "overall_df": pd.DataFrame(),
                "importances_df": pd.DataFrame()
            }
            continue
        
        # Ensure numeric cols
        num_cols = [
            "infection_rate_selected", "infection_rate_non_selected",
            "npi_level_selected", "npi_level_non_selected"
        ]
        
        if "cap_z" in features:
            num_cols.append("capacity_context")
        if "income_loss_diff_z" in features:
            num_cols.append("income_loss_selected")
            num_cols.append("income_loss_non_selected")
        if "preferred_option_num_z" in features:
            num_cols.append("preferred_option_num")
        
        for c in num_cols:
            if c in df.columns:
                df[c] = pd.to_numeric(df[c], errors="coerce")
        
        # Choice & attribute reconstruction
        df["choice"] = (df["selected_choice"] == df["pair"].str.split("-").str[0]).astype(int)
        df["inf_1"] = np.where(df["choice"] == 1, df["infection_rate_selected"], df["infection_rate_non_selected"])
        df["inf_2"] = np.where(df["choice"] == 1, df["infection_rate_non_selected"], df["infection_rate_selected"])
        df["npi_1"] = np.where(df["choice"] == 1, df["npi_level_selected"], df["npi_level_non_selected"])
        df["npi_2"] = np.where(df["choice"] == 1, df["npi_level_non_selected"], df["npi_level_selected"])
        
        df["bc_diff"] = df["npi_1"] - df["npi_2"]
        df["inf_diff"] = df["inf_1"] - df["inf_2"]
        
        # Fit per model
        per_model_rows = []
        importances_rows = []
        pooled_y = []
        pooled_score = []
        
        models_sorted = sorted(df["model"].dropna().unique())
        
        for model_name in models_sorted:
            m_df = df[df["model"] == model_name].copy()
            
            # Build required columns list for dropna
            required_cols = ["choice", "bc_diff", "inf_diff"]
            if "cap_z" in features:
                required_cols.append("capacity_context")
            if "income_loss_diff_z" in features:
                required_cols.append("income_loss_selected")
                required_cols.append("income_loss_non_selected")
            if "preferred_option_num_z" in features:
                required_cols.append("preferred_option_num")
            
            m_df = m_df.dropna(subset=required_cols)
            
            if len(m_df) < min_obs:
                continue
            
            m_df = add_features(m_df, features)
            if len(m_df) < min_obs:
                continue
            
            X = m_df[features].to_numpy(dtype=float)
            y = m_df["choice"].to_numpy(dtype=int)
            
            # Must have both classes
            vals, counts = np.unique(y, return_counts=True)
            if len(vals) < 2:
                continue
            
            # Folds can't exceed minority class count
            min_class = int(np.min(counts))
            n_splits = int(min(cv_folds, min_class))
            if n_splits < 2:
                continue
            
            # CV predicted probabilities for AUROC
            cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=random_state)
            y_score_cv = np.full_like(y, fill_value=np.nan, dtype=float)
            
            for fold_i, (tr, te) in enumerate(cv.split(X, y)):
                rf = RandomForestClassifier(
                    n_estimators=n_estimators,
                    max_depth=max_depth,
                    min_samples_leaf=min_samples_leaf,
                    random_state=random_state + fold_i,
                    n_jobs=n_jobs
                )
                rf.fit(X[tr], y[tr])
                y_score_cv[te] = rf.predict_proba(X[te])[:, 1]
            
            auroc_cv = float(roc_auc_score(y, y_score_cv))
            
            pooled_y.append(y)
            pooled_score.append(y_score_cv)
            
            # Fit final RF on full data for feature_importances_
            rf_full = RandomForestClassifier(
                n_estimators=n_estimators,
                max_depth=max_depth,
                min_samples_leaf=min_samples_leaf,
                random_state=random_state,
                n_jobs=n_jobs
            )
            rf_full.fit(X, y)
            importances = pd.Series(rf_full.feature_importances_, index=features).sort_values(ascending=False)
            
            per_model_rows.append({
                "Model": model_name,
                "N": int(len(m_df)),
                "CV_folds_used": int(n_splits),
                "AUROC_CV": auroc_cv
            })
            
            for feat, imp in importances.items():
                importances_rows.append({
                    "Model": model_name,
                    "feature": feat,
                    "importance": float(imp)
                })
        
        per_model_df = pd.DataFrame(per_model_rows).sort_values("Model").reset_index(drop=True)
        importances_df = pd.DataFrame(importances_rows)
        
        # Overall pooled AUROC
        overall_df = pd.DataFrame([{
            "Strategy": strategy_name,
            "Models_included": int(per_model_df["Model"].nunique()) if not per_model_df.empty else 0,
            "Total_N": int(per_model_df["N"].sum()) if not per_model_df.empty else 0,
            "AUROC_CV_pooled": (
                float(roc_auc_score(np.concatenate(pooled_y), np.concatenate(pooled_score)))
                if len(pooled_y) > 0 else np.nan
            )
        }])
        
        show_df(per_model_df, f"Random Forest: {strategy_name} per-model CV AUROC")
        show_df(overall_df, f"Random Forest: {strategy_name} pooled CV AUROC")
        
        all_results[strategy_name] = {
            "per_model_df": per_model_df,
            "overall_df": overall_df,
            "importances_df": importances_df
        }
    
    
    # --- Visualization: 2x2 grid ---
    if plot:
        fig, axes = plt.subplots(2, 2, figsize=FIG_SIZE_MULTI)
        axes = axes.ravel()
        
        # Create legend handles
        legend_handles = [plt.Rectangle((0, 0), 1, 1, fc=MODEL_COLORS[model], alpha=0.85) 
                         for model in MODEL_ORDER]
        
        for idx, (strategy_name, features) in enumerate(strategy_features.items()):
            ax = axes[idx]
            importances_df = all_results[strategy_name]["importances_df"]
            
            if importances_df.empty:
                ax.text(0.5, 0.5, 'No Data', ha='center', va='center', fontsize=20)
                ax.set_title(strategy_titles[strategy_name], fontweight='bold', pad=10)
                ax.axis('off')
                continue
            
            # Map model names to display names
            importances_df['Model_Display'] = importances_df['Model'].map(MODEL_NAME_MAPPING)
            importances_df['Model_Display'] = importances_df['Model_Display'].fillna(importances_df['Model'])
            
            # Pivot for grouped bar chart
            pivot_df = importances_df.pivot(index='feature', columns='Model_Display', values='importance')
            
            # Reorder columns (models) alphabetically: ChatGPT, DeepSeek, Gemini, Grok
            available_models = [m for m in MODEL_ORDER if m in pivot_df.columns]
            pivot_df = pivot_df[available_models]
            
            # Reorder rows (features) in the order they appear in the feature list
            available_features = [f for f in features if f in pivot_df.index]
            pivot_df = pivot_df.reindex(available_features)
            
            # Set bar width and positions
            n_features = len(pivot_df)
            n_models = len(pivot_df.columns)
            bar_height = 0.15
            y_positions = np.arange(n_features)
            
            # Plot bars for each model
            for i, model in enumerate(pivot_df.columns):
                offset = (i - n_models/2 + 0.5) * bar_height
                color = MODEL_COLORS.get(model, 'gray')
                ax.barh(y_positions + offset, pivot_df[model], 
                       height=bar_height, label=model, color=color, alpha=0.85)
            
            # Customize plot
            ax.set_yticks(y_positions)
            ax.set_yticklabels(pivot_df.index, fontsize=14)
            ax.set_xlabel('Feature Importance', fontweight='bold', fontsize=14)
            ax.set_title(strategy_titles[strategy_name], fontweight='bold', pad=10, fontsize=16)
            
            # Grid
            ax.grid(True, axis='x', alpha=0.3, linestyle='--')
            ax.set_axisbelow(True)
        
        # Overall title
        fig.suptitle('Random Forest Feature Importances Across Strategies', 
                    fontweight='bold', fontsize=20, y=0.98)
        
        # Add centered horizontal legend below title
        fig.legend(legend_handles, MODEL_ORDER, 
                  loc='upper center', 
                  ncol=4, 
                  frameon=True, 
                  fancybox=True, 
                  shadow=True,
                  fontsize=14,
                  bbox_to_anchor=(0.5, 0.95))
        
        plt.tight_layout(rect=[0, 0, 1, 0.94])
        plt.show()
    
    return all_results


# Example usage:
# all_results = run_rf_all_strategies_comparison(full_results_df)
```

# Elastic-Net v2 All Strategy

## Function

```python
def run_elasticnet_all_strategies_with_interactions(
    results_df,
    min_obs=10,
    show_tables=True,
    plot=True,
    # Elastic Net params
    cv_folds=10,
    l1_ratios=np.linspace(0.1, 1.0, 10),
    alphas=np.logspace(-4, 1, 50),
    random_state=42,
    n_jobs=-1,
    # Bootstrap params for CIs
    n_bootstrap=1000,
    ci_percentile=95,
    # Feature selection params
    perm_importance_threshold=0.02,
    n_perm_repeats=10
):
    """
    Elastic-Net Logistic Regression for all 4 strategies with feature interactions.
    Computed separately per LLM model with model-specific feature selection.
    
    Strategy 1: inf_diff_z, bc_diff_z (and interactions)
    Strategy 2: inf_diff_z, bc_diff_z, cap_z (and interactions)
    Strategy 3: inf_diff_z, bc_diff_z, cap_z, income_loss_diff_z (and interactions)
    Strategy 4: inf_diff_z, bc_diff_z, cap_z, income_loss_diff_z, preferred_option_num_z (and interactions)
    
    Note: cap_z is excluded as a main effect (option-independent context variable).
          All other interactions with option-dependent features are included.
    
    Returns:
      - all_results: dict with results for each strategy
      - 2x2 grid of forest plots
      - AUROC summary table
    """
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import seaborn as sns
    from sklearn.linear_model import LogisticRegressionCV, LogisticRegression
    from sklearn.preprocessing import StandardScaler
    from sklearn.model_selection import StratifiedKFold
    from sklearn.inspection import permutation_importance
    from sklearn.metrics import roc_auc_score
    from sklearn.utils import resample
    from itertools import combinations
    import warnings
    warnings.filterwarnings('ignore')
    
    
    # --- Styling Configuration ---
    MODEL_NAME_MAPPING = {
        'x-ai/grok-4-fast': 'Grok',
        'openai/gpt-5-nano': 'ChatGPT',
        'google/gemini-2.5-flash': 'Gemini',
        'deepseek/deepseek-chat-v3.1': 'DeepSeek'
    }
    
    MODEL_ORDER = ['ChatGPT', 'DeepSeek', 'Gemini', 'Grok']
    _palette_raw = sns.color_palette("tab10", n_colors=len(MODEL_ORDER))
    MODEL_COLORS = dict(zip(MODEL_ORDER, _palette_raw))
    
    FIG_SIZE_MULTI = (22, 20)
    FONT_SCALE = 2.2
    DPI = 600
    
    sns.set_style("white")
    sns.set_context("paper", font_scale=FONT_SCALE)
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']
    plt.rcParams['figure.dpi'] = DPI
    
    
    # --- Strategy definitions ---
    strategy_base_features = {
        "strategy_one": ["inf_diff_z", "bc_diff_z"],
        "strategy_two": ["inf_diff_z", "bc_diff_z", "cap_z"],
        "strategy_three": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z"],
        "strategy_four": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z", "preferred_option_num_z"]
    }
    
    strategy_titles = {
        "strategy_one": "Strategy 1",
        "strategy_two": "Strategy 2",
        "strategy_three": "Strategy 3",
        "strategy_four": "Strategy 4"
    }
    
    
    # --- Helper: NumPy z-score ---
    def np_zscore(series: pd.Series) -> pd.Series:
        arr = np.asarray(series, dtype=float)
        mu = np.nanmean(arr)
        sigma = np.nanstd(arr, ddof=0)
        if (not np.isfinite(sigma)) or sigma == 0:
            return pd.Series(np.full_like(arr, np.nan, dtype=float), index=series.index)
        return pd.Series((arr - mu) / sigma, index=series.index)
    
    
    # --- Display helper ---
    def show_df(df, title):
        if not show_tables:
            return
        print("\n" + "=" * 80)
        print(title)
        print("=" * 80)
        try:
            from IPython.display import display
            display(df)
        except Exception:
            print(df.to_string(index=False))
    
    
    # --- Create polynomial interaction features ---
    def create_interaction_features(df, base_features):
        """
        Create interaction features without repeated terms.
        Excludes:
        - cap_z as main effect (option-independent context variable)
        - Any purely option-independent combinations
        
        Includes:
        - All interactions with at least one option-dependent feature
          (inf_diff_z, bc_diff_z, income_loss_diff_z, preferred_option_num_z)
        """
        interaction_df = df[base_features].copy()
        feature_names = list(base_features)
        
        # Define option-independent features (exclude as main effects)
        option_independent = {'cap_z'}
        # Define option-dependent features
        option_dependent = {'inf_diff_z', 'bc_diff_z', 'income_loss_diff_z', 'preferred_option_num_z'}
        
        # Max degree is the number of base features
        max_degree = len(base_features)
        
        # Generate interactions from degree 2 to max_degree
        for degree in range(2, max_degree + 1):
            for combo in combinations(base_features, degree):
                combo_set = set(combo)
                
                # Skip if interaction is purely option-independent
                if combo_set.issubset(option_independent):
                    continue
                
                # Skip if interaction doesn't include at least one option-dependent feature
                if not combo_set.intersection(option_dependent):
                    continue
                
                feature_name = '_x_'.join(combo)
                
                # Compute interaction
                interaction_df[feature_name] = df[combo[0]]
                for feat in combo[1:]:
                    interaction_df[feature_name] *= df[feat]
                
                feature_names.append(feature_name)
        
        return interaction_df, feature_names
    
    
    # --- Add base features ---
    def add_base_features(m_df: pd.DataFrame, strategy_name: str) -> pd.DataFrame:
        m_df = m_df.copy()
        
        # Always z-score inf_diff and bc_diff
        m_df["inf_diff_z"] = np_zscore(m_df["inf_diff"])
        m_df["bc_diff_z"] = np_zscore(m_df["bc_diff"])
        
        # Add cap_z for strategies 2, 3, 4
        if strategy_name in ["strategy_two", "strategy_three", "strategy_four"]:
            m_df["cap_z"] = np_zscore(m_df["capacity_context"])
        
        # Add income_loss_diff_z for strategies 3, 4
        if strategy_name in ["strategy_three", "strategy_four"]:
            # Exogenous income loss differential: option_1 minus option_2, independent of choice
            m_df["income_loss_1"] = np.where(m_df["choice"] == 1, m_df["income_loss_selected"], m_df["income_loss_non_selected"])
            m_df["income_loss_2"] = np.where(m_df["choice"] == 1, m_df["income_loss_non_selected"], m_df["income_loss_selected"])
            m_df["income_loss_diff"] = m_df["income_loss_1"] - m_df["income_loss_2"]
            m_df["income_loss_diff_z"] = np_zscore(m_df["income_loss_diff"])
        
        # Add preferred_option_num_z for strategy 4
        if strategy_name == "strategy_four":
            m_df["preferred_option_num_z"] = np_zscore(m_df["preferred_option_num"])
        
        return m_df
    
    
    # --- Bootstrap CI calculation ---
    def bootstrap_coefficients(X, y, C_opt, l1_ratio_opt, n_bootstrap, random_state):
        """Bootstrap confidence intervals for coefficients."""
        np.random.seed(random_state)
        coef_samples = []
        
        for i in range(n_bootstrap):
            X_boot, y_boot = resample(X, y, random_state=random_state + i)
            
            if len(np.unique(y_boot)) < 2:
                continue
            
            model = LogisticRegression(
                penalty='elasticnet',
                solver='saga',
                C=C_opt,
                l1_ratio=l1_ratio_opt,
                max_iter=10000,
                random_state=random_state + i
            )
            
            try:
                model.fit(X_boot, y_boot)
                coef_samples.append(model.coef_[0])
            except:
                continue
        
        return np.array(coef_samples)
    
    
    # --- Process each strategy ---
    all_results = {}
    auroc_summary_rows = []
    
    for strategy_name, base_features in strategy_base_features.items():
        print(f"\n{'='*80}")
        print(f"Processing {strategy_name}: {base_features}")
        print(f"{'='*80}")
        
        # Filter strategy
        df = results_df[results_df["strategy"] == strategy_name].copy()
        print(f"Found {len(df)} rows for {strategy_name}")
        
        if df.empty:
            all_results[strategy_name] = {
                "per_model_df": pd.DataFrame(),
                "overall_df": pd.DataFrame(),
                "odds_ratios_df": pd.DataFrame()
            }
            continue
        
        # Ensure numeric columns
        num_cols = [
            "infection_rate_selected", "infection_rate_non_selected",
            "npi_level_selected", "npi_level_non_selected"
        ]
        
        if "cap_z" in base_features:
            num_cols.append("capacity_context")
        if "income_loss_diff_z" in base_features:
            num_cols.append("income_loss_selected")
            num_cols.append("income_loss_non_selected")
        if "preferred_option_num_z" in base_features:
            num_cols.append("preferred_option_num")
        
        for c in num_cols:
            if c in df.columns:
                df[c] = pd.to_numeric(df[c], errors="coerce")
        
        # Reconstruct choice and differences
        df["choice"] = (df["selected_choice"] == df["pair"].str.split("-").str[0]).astype(int)
        df["inf_1"] = np.where(df["choice"] == 1, df["infection_rate_selected"], df["infection_rate_non_selected"])
        df["inf_2"] = np.where(df["choice"] == 1, df["infection_rate_non_selected"], df["infection_rate_selected"])
        df["npi_1"] = np.where(df["choice"] == 1, df["npi_level_selected"], df["npi_level_non_selected"])
        df["npi_2"] = np.where(df["choice"] == 1, df["npi_level_non_selected"], df["npi_level_selected"])
        
        df["bc_diff"] = df["npi_1"] - df["npi_2"]
        df["inf_diff"] = df["inf_1"] - df["inf_2"]
        
        # Add base features
        df = add_base_features(df, strategy_name)
        
        # Build required columns for dropna
        required_cols = ["choice", "inf_diff", "bc_diff"]
        if "cap_z" in base_features:
            required_cols.append("capacity_context")
        if "income_loss_diff_z" in base_features:
            required_cols.append("income_loss_selected")
            required_cols.append("income_loss_non_selected")
        if "preferred_option_num_z" in base_features:
            required_cols.append("preferred_option_num")
        
        df = df.dropna(subset=required_cols)
        
        if len(df) < min_obs:
            print(f"Insufficient data after cleaning: {len(df)} rows")
            all_results[strategy_name] = {
                "per_model_df": pd.DataFrame(),
                "overall_df": pd.DataFrame(),
                "odds_ratios_df": pd.DataFrame()
            }
            continue
        
        # Create interaction features
        print(f"Creating interaction features...")
        interaction_df, all_feature_names = create_interaction_features(df, base_features)
        print(f"Created {len(all_feature_names)} features (including interactions)")
        
        # --- Permutation Importance Feature Selection per Model ---
        print(f"Performing permutation importance analysis per model...")
        
        models_sorted = sorted(df["model"].dropna().unique())
        perm_importance_by_model = {model: {feat: None for feat in all_feature_names} for model in models_sorted}
        features_important_to_any_model = set()
        
        for model_name in models_sorted:
            m_indices = df["model"] == model_name
            X_model = interaction_df.loc[m_indices, all_feature_names].to_numpy(dtype=float)
            y_model = df.loc[m_indices, "choice"].to_numpy(dtype=int)
            
            if len(y_model) < min_obs or len(np.unique(y_model)) < 2:
                continue
            
            lr_temp = LogisticRegression(penalty='l2', max_iter=1000, random_state=random_state)
            lr_temp.fit(X_model, y_model)
            
            perm_result = permutation_importance(
                lr_temp, X_model, y_model,
                n_repeats=n_perm_repeats,
                random_state=random_state,
                n_jobs=n_jobs
            )
            
            for idx, feat in enumerate(all_feature_names):
                importance = perm_result.importances_mean[idx]
                perm_importance_by_model[model_name][feat] = importance
                
                if importance >= perm_importance_threshold:
                    features_important_to_any_model.add(feat)
        
        # Remove option-independent features as main effects
        features_important_to_any_model = {
            f for f in features_important_to_any_model 
            if f not in ['cap_z']
        }
        
        print(f"Features important to ANY model: {len(features_important_to_any_model)}")
        
        if len(features_important_to_any_model) == 0:
            print(f"No features passed threshold")
            all_results[strategy_name] = {
                "per_model_df": pd.DataFrame(),
                "overall_df": pd.DataFrame(),
                "odds_ratios_df": pd.DataFrame()
            }
            continue
        
        # Create per-model feature sets
        model_feature_sets = {}
        for model_name in models_sorted:
            model_features = [
                feat for feat in features_important_to_any_model
                if perm_importance_by_model[model_name][feat] is not None 
                and perm_importance_by_model[model_name][feat] >= perm_importance_threshold
            ]
            model_feature_sets[model_name] = model_features
        
        # --- Process Each Model ---
        per_model_rows = []
        odds_ratios_rows = []
        pooled_y = []
        pooled_score = []
        
        for model_name in models_sorted:
            selected_features = model_feature_sets.get(model_name, [])
            
            if len(selected_features) == 0:
                continue
            
            m_df = df[df["model"] == model_name].copy()
            
            if len(m_df) < min_obs:
                continue
            
            X = interaction_df.loc[m_df.index, selected_features].to_numpy(dtype=float)
            y = m_df["choice"].to_numpy(dtype=int)
            
            vals, counts = np.unique(y, return_counts=True)
            if len(vals) < 2:
                continue
            
            min_class = int(np.min(counts))
            n_splits = int(min(cv_folds, min_class))
            if n_splits < 2:
                continue
            
            # Standardize features
            scaler = StandardScaler()
            X_scaled = scaler.fit_transform(X)
            
            # --- Hyperparameter Optimization ---
            elasticnet_model = LogisticRegressionCV(
                Cs=1.0/alphas,
                cv=n_splits,
                penalty='elasticnet',
                solver='saga',
                l1_ratios=l1_ratios,
                max_iter=10000,
                random_state=random_state,
                n_jobs=n_jobs,
                scoring='roc_auc'
            )
            
            try:
                elasticnet_model.fit(X_scaled, y)
            except Exception as e:
                continue
            
            C_opt = elasticnet_model.C_[0]
            alpha_opt = 1.0 / C_opt
            l1_ratio_opt = elasticnet_model.l1_ratio_[0]
            
            coefficients = elasticnet_model.coef_[0]
            
            # --- Calculate CV AUROC ---
            cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=random_state)
            y_pred_proba = np.zeros(len(y))
            
            for fold_i, (train_idx, test_idx) in enumerate(cv.split(X_scaled, y)):
                model_cv = LogisticRegression(
                    penalty='elasticnet',
                    solver='saga',
                    C=C_opt,
                    l1_ratio=l1_ratio_opt,
                    max_iter=10000,
                    random_state=random_state + fold_i
                )
                model_cv.fit(X_scaled[train_idx], y[train_idx])
                y_pred_proba[test_idx] = model_cv.predict_proba(X_scaled[test_idx])[:, 1]
            
            auroc_cv = roc_auc_score(y, y_pred_proba)
            
            pooled_y.append(y)
            pooled_score.append(y_pred_proba)
            
            # Store AUROC for summary table
            auroc_summary_rows.append({
                "Strategy": strategy_titles[strategy_name],
                "Model": model_name,
                "AUROC_CV": auroc_cv,
                "N": len(m_df)
            })
            
            # --- Bootstrap Confidence Intervals ---
            coef_samples = bootstrap_coefficients(
                X_scaled, y, C_opt, l1_ratio_opt,
                n_bootstrap, random_state
            )
            
            lower_percentile = (100 - ci_percentile) / 2
            upper_percentile = 100 - lower_percentile
            
            coef_lower = np.percentile(coef_samples, lower_percentile, axis=0)
            coef_upper = np.percentile(coef_samples, upper_percentile, axis=0)
            
            ors = np.exp(coefficients)
            or_lower = np.exp(coef_lower)
            or_upper = np.exp(coef_upper)
            
            per_model_rows.append({
                "Model": model_name,
                "N": int(len(m_df)),
                "N_features": len(selected_features),
                "CV_folds_used": int(n_splits),
                "AUROC_CV": auroc_cv,
                "optimal_alpha": alpha_opt,
                "optimal_l1_ratio": l1_ratio_opt
            })
            
            for feat, coef, or_val, ci_low, ci_high in zip(
                selected_features, coefficients, ors, or_lower, or_upper
            ):
                odds_ratios_rows.append({
                    "Model": model_name,
                    "feature": feat,
                    "coefficient": float(coef),
                    "odds_ratio": float(or_val),
                    "ci_lower": float(ci_low),
                    "ci_upper": float(ci_high)
                })
        
        # Create results dataframes
        per_model_df = pd.DataFrame(per_model_rows).sort_values("Model").reset_index(drop=True)
        odds_ratios_df = pd.DataFrame(odds_ratios_rows)
        
        overall_df = pd.DataFrame([{
            "Strategy": strategy_name,
            "Models_included": int(per_model_df["Model"].nunique()) if not per_model_df.empty else 0,
            "Total_N": int(per_model_df["N"].sum()) if not per_model_df.empty else 0,
            "AUROC_CV_pooled": (
                float(roc_auc_score(np.concatenate(pooled_y), np.concatenate(pooled_score)))
                if len(pooled_y) > 0 else np.nan
            )
        }])
        
        show_df(per_model_df, f"Elastic-Net: {strategy_name} per-model results")
        show_df(odds_ratios_df.head(20), f"Elastic-Net: {strategy_name} odds ratios (first 20)")
        
        all_results[strategy_name] = {
            "per_model_df": per_model_df,
            "overall_df": overall_df,
            "odds_ratios_df": odds_ratios_df,
            "features_important_to_any_model": sorted(features_important_to_any_model),
            "model_feature_sets": model_feature_sets
        }
    
    # --- AUROC Summary Table ---
    auroc_summary_df = pd.DataFrame(auroc_summary_rows)
    if not auroc_summary_df.empty:
        # Pivot to create table with strategies as rows and models as columns
        auroc_pivot = auroc_summary_df.pivot(index='Strategy', columns='Model', values='AUROC_CV')
        
        # Map model names and reorder
        auroc_pivot.columns = auroc_pivot.columns.map(lambda x: MODEL_NAME_MAPPING.get(x, x))
        available_models = [m for m in MODEL_ORDER if m in auroc_pivot.columns]
        auroc_pivot = auroc_pivot[available_models]
        
        # Reorder strategies
        strategy_order = ['Strategy 1', 'Strategy 2', 'Strategy 3', 'Strategy 4']
        auroc_pivot = auroc_pivot.reindex([s for s in strategy_order if s in auroc_pivot.index])
        
        show_df(auroc_pivot.round(4), "AUROC Summary: Cross-Validated AUROC by Strategy and Model")
    
    # --- Visualization: 2x2 grid ---
    if plot:
        fig, axes = plt.subplots(2, 2, figsize=FIG_SIZE_MULTI)
        axes = axes.ravel()
        
        # Create legend handles
        all_models_in_data = set()
        for strategy_name in strategy_base_features.keys():
            if not all_results[strategy_name]["odds_ratios_df"].empty:
                models = all_results[strategy_name]["odds_ratios_df"]["Model"].unique()
                all_models_in_data.update(models)
        
        all_models_display = [MODEL_NAME_MAPPING.get(m, m) for m in all_models_in_data]
        available_legend_models = [m for m in MODEL_ORDER if m in all_models_display]
        legend_handles = [plt.Rectangle((0, 0), 1, 1, fc=MODEL_COLORS[model], alpha=0.85) 
                         for model in available_legend_models]
        
        for idx, (strategy_name, base_features) in enumerate(strategy_base_features.items()):
            ax = axes[idx]
            odds_ratios_df = all_results[strategy_name]["odds_ratios_df"]
            
            if odds_ratios_df.empty:
                ax.text(0.5, 0.5, 'No Data', ha='center', va='center', fontsize=24)
                ax.set_title(strategy_titles[strategy_name], fontweight='bold', pad=15, fontsize=20)
                ax.axis('off')
                continue
            
            # Map model names
            odds_ratios_df = odds_ratios_df.copy()
            odds_ratios_df['Model_Display'] = odds_ratios_df['Model'].map(MODEL_NAME_MAPPING)
            odds_ratios_df['Model_Display'] = odds_ratios_df['Model_Display'].fillna(odds_ratios_df['Model'])
            
            # Pivot
            pivot_or = odds_ratios_df.pivot(index='feature', columns='Model_Display', values='odds_ratio')
            pivot_ci_low = odds_ratios_df.pivot(index='feature', columns='Model_Display', values='ci_lower')
            pivot_ci_high = odds_ratios_df.pivot(index='feature', columns='Model_Display', values='ci_upper')
            
            # Reorder columns (models)
            available_models = [m for m in MODEL_ORDER if m in pivot_or.columns]
            pivot_or = pivot_or[available_models]
            pivot_ci_low = pivot_ci_low[available_models]
            pivot_ci_high = pivot_ci_high[available_models]
            
            # Reorder rows (features)
            features_important = all_results[strategy_name].get("features_important_to_any_model", [])
            available_features = [f for f in features_important if f in pivot_or.index]
            pivot_or = pivot_or.reindex(available_features)
            pivot_ci_low = pivot_ci_low.reindex(available_features)
            pivot_ci_high = pivot_ci_high.reindex(available_features)
            
            # Set positions
            n_features = len(pivot_or)
            n_models = len(pivot_or.columns)
            
            if n_features == 0:
                ax.text(0.5, 0.5, 'No Features', ha='center', va='center', fontsize=20)
                ax.set_title(strategy_titles[strategy_name], fontweight='bold', pad=15, fontsize=20)
                ax.axis('off')
                continue
            
            spacing = 0.18
            y_positions = np.arange(n_features) * 1.2
            
            # Plot dots with error bars
            for i, model in enumerate(pivot_or.columns):
                offset = (i - n_models/2 + 0.5) * spacing
                color = MODEL_COLORS.get(model, 'gray')
                
                or_vals = pivot_or[model].values
                ci_low_vals = pivot_ci_low[model].values
                ci_high_vals = pivot_ci_high[model].values
                
                for j in range(n_features):
                    if pd.notna(or_vals[j]):
                        or_val = or_vals[j]
                        ci_low = ci_low_vals[j] if pd.notna(ci_low_vals[j]) else or_val
                        ci_high = ci_high_vals[j] if pd.notna(ci_high_vals[j]) else or_val
                        
                        left_err = max(0.0, or_val - ci_low)
                        right_err = max(0.0, ci_high - or_val)
                        
                        ax.errorbar(
                            or_val, y_positions[j] + offset,
                            xerr=[[left_err], [right_err]],
                            fmt='o',
                            color=color,
                            ecolor=color,
                            markersize=10,
                            capsize=5,
                            alpha=0.85,
                            linewidth=2.5
                        )
            
            # Add vertical line at OR=1
            ax.axvline(x=1, color='black', linestyle='--', linewidth=2, alpha=0.7, zorder=1)
            
            # Customize plot with larger fonts
            ax.set_yticks(y_positions)
            ax.set_yticklabels(pivot_or.index, fontsize=16)
            ax.set_xlabel('Odds Ratio (95% CI)', fontweight='bold', fontsize=18)
            ax.set_title(strategy_titles[strategy_name], fontweight='bold', pad=15, fontsize=20)
            
            # Increase x-axis tick label size
            ax.tick_params(axis='x', labelsize=16)
            ax.tick_params(axis='y', labelsize=16)
            
            # Use log scale if ORs span wide range
            all_or_vals = pivot_or.values.flatten()
            all_or_vals = all_or_vals[~np.isnan(all_or_vals)]
            if len(all_or_vals) > 0 and (np.max(all_or_vals) / np.min(all_or_vals) > 10):
                ax.set_xscale('log')
            
            # Grid
            ax.grid(True, axis='x', alpha=0.3, linestyle='--', linewidth=1.5)
            ax.set_axisbelow(True)
        
        # Overall title
        fig.suptitle('Elastic-Net Regression: Odds Ratios with 95% CI Across Strategies', 
                    fontweight='bold', fontsize=24, y=0.995)
        
        # Add legend with larger font
        fig.legend(legend_handles, available_legend_models, 
                  loc='upper center', 
                  ncol=4, 
                  frameon=True, 
                  fancybox=True, 
                  shadow=True,
                  fontsize=16,
                  bbox_to_anchor=(0.5, 0.97))
        
        plt.tight_layout(rect=[0, 0, 1, 0.96])
        plt.show()
    
    # --- Return results ---
    return {
        'all_results': all_results,
        'auroc_summary': auroc_pivot if not auroc_summary_df.empty else pd.DataFrame()
    }
```

```python
def run_rf_all_strategies_analysis(
    results_df,
    min_obs=10,
    show_tables=True,
    # Random Forest params
    n_estimators=500,
    max_depth=None,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1,
    # CV params
    cv_folds=10
):
    """
    Random Forest classifier for all 4 strategies with strategy-specific features.
    
    Strategy 1: inf_diff_z, bc_diff_z
    Strategy 2: inf_diff_z, bc_diff_z, cap_z
    Strategy 3: inf_diff_z, bc_diff_z, cap_z, income_loss_diff_z
    Strategy 4: inf_diff_z, bc_diff_z, cap_z, income_loss_diff_z, preferred_option_num_z
    
    Outputs:
      - all_results: dict with per_model_df, overall_df, importances_df for each strategy
    """
    import pandas as pd
    import numpy as np
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.metrics import roc_auc_score
    from sklearn.model_selection import StratifiedKFold
    
    
    # --- Helper: NumPy z-score ---
    def np_zscore(series: pd.Series) -> pd.Series:
        arr = np.asarray(series, dtype=float)
        mu = np.nanmean(arr)
        sigma = np.nanstd(arr, ddof=0)
        if (not np.isfinite(sigma)) or sigma == 0:
            return pd.Series(np.full_like(arr, np.nan, dtype=float), index=series.index)
        return pd.Series((arr - mu) / sigma, index=series.index)
    
    
    # --- Display helper ---
    def show_df(df, title):
        if not show_tables:
            return
        print("\n" + "=" * 60)
        print(title)
        print("=" * 60)
        try:
            from IPython.display import display
            display(df)
        except Exception:
            print(df.to_string(index=False))
    
    
    # --- Strategy-specific feature definitions (in order) ---
    strategy_features = {
        "strategy_one": ["inf_diff_z", "bc_diff_z"],
        "strategy_two": ["inf_diff_z", "bc_diff_z", "cap_z"],
        "strategy_three": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z"],
        "strategy_four": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z", "preferred_option_num_z"]
    }
    
    
    def add_features(m_df: pd.DataFrame, features: list) -> pd.DataFrame:
        m_df = m_df.copy()
        
        # Z-score all potential features
        m_df["inf_diff_z"] = np_zscore(m_df["inf_diff"])
        m_df["bc_diff_z"]  = np_zscore(m_df["bc_diff"])
        
        if "cap_z" in features:
            m_df["cap_z"] = np_zscore(m_df["capacity_context"])
        
        if "income_loss_diff_z" in features:
            # Exogenous income loss differential: option_1 minus option_2, independent of choice
            m_df["income_loss_1"] = np.where(m_df["choice"] == 1, m_df["income_loss_selected"], m_df["income_loss_non_selected"])
            m_df["income_loss_2"] = np.where(m_df["choice"] == 1, m_df["income_loss_non_selected"], m_df["income_loss_selected"])
            m_df["income_loss_diff"] = m_df["income_loss_1"] - m_df["income_loss_2"]
            m_df["income_loss_diff_z"] = np_zscore(m_df["income_loss_diff"])
        
        if "preferred_option_num_z" in features:
            m_df["preferred_option_num_z"] = np_zscore(m_df["preferred_option_num"])
        
        return m_df.dropna(subset=["choice"] + features)
    
    
    # --- Process each strategy ---
    all_results = {}
    
    for strategy_name, features in strategy_features.items():
        print(f"\n{'='*60}")
        print(f"Processing {strategy_name}")
        print(f"{'='*60}")
        
        # Filter strategy
        df = results_df[results_df["strategy"] == strategy_name].copy()
        print(f"Found {len(df)} rows for {strategy_name}")
        
        if df.empty:
            all_results[strategy_name] = {
                "per_model_df": pd.DataFrame(),
                "overall_df": pd.DataFrame(),
                "importances_df": pd.DataFrame()
            }
            continue
        
        # Ensure numeric cols
        num_cols = [
            "infection_rate_selected", "infection_rate_non_selected",
            "npi_level_selected", "npi_level_non_selected"
        ]
        
        if "cap_z" in features:
            num_cols.append("capacity_context")
        if "income_loss_diff_z" in features:
            num_cols.append("income_loss_selected")
            num_cols.append("income_loss_non_selected")
        if "preferred_option_num_z" in features:
            num_cols.append("preferred_option_num")
        
        for c in num_cols:
            if c in df.columns:
                df[c] = pd.to_numeric(df[c], errors="coerce")
        
        # Choice & attribute reconstruction
        df["choice"] = (df["selected_choice"] == df["pair"].str.split("-").str[0]).astype(int)
        df["inf_1"] = np.where(df["choice"] == 1, df["infection_rate_selected"], df["infection_rate_non_selected"])
        df["inf_2"] = np.where(df["choice"] == 1, df["infection_rate_non_selected"], df["infection_rate_selected"])
        df["npi_1"] = np.where(df["choice"] == 1, df["npi_level_selected"], df["npi_level_non_selected"])
        df["npi_2"] = np.where(df["choice"] == 1, df["npi_level_non_selected"], df["npi_level_selected"])
        
        df["bc_diff"] = df["npi_1"] - df["npi_2"]
        df["inf_diff"] = df["inf_1"] - df["inf_2"]
        
        # Fit per model
        per_model_rows = []
        importances_rows = []
        pooled_y = []
        pooled_score = []
        
        models_sorted = sorted(df["model"].dropna().unique())
        
        for model_name in models_sorted:
            m_df = df[df["model"] == model_name].copy()
            
            # Build required columns list for dropna
            required_cols = ["choice", "bc_diff", "inf_diff"]
            if "cap_z" in features:
                required_cols.append("capacity_context")
            if "income_loss_diff_z" in features:
                required_cols.append("income_loss_selected")
                required_cols.append("income_loss_non_selected")
            if "preferred_option_num_z" in features:
                required_cols.append("preferred_option_num")
            
            m_df = m_df.dropna(subset=required_cols)
            
            if len(m_df) < min_obs:
                continue
            
            m_df = add_features(m_df, features)
            if len(m_df) < min_obs:
                continue
            
            X = m_df[features].to_numpy(dtype=float)
            y = m_df["choice"].to_numpy(dtype=int)
            
            # Must have both classes
            vals, counts = np.unique(y, return_counts=True)
            if len(vals) < 2:
                continue
            
            # Folds can't exceed minority class count
            min_class = int(np.min(counts))
            n_splits = int(min(cv_folds, min_class))
            if n_splits < 2:
                continue
            
            # CV predicted probabilities for AUROC
            cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=random_state)
            y_score_cv = np.full_like(y, fill_value=np.nan, dtype=float)
            
            for fold_i, (tr, te) in enumerate(cv.split(X, y)):
                rf = RandomForestClassifier(
                    n_estimators=n_estimators,
                    max_depth=max_depth,
                    min_samples_leaf=min_samples_leaf,
                    random_state=random_state + fold_i,
                    n_jobs=n_jobs
                )
                rf.fit(X[tr], y[tr])
                y_score_cv[te] = rf.predict_proba(X[te])[:, 1]
            
            auroc_cv = float(roc_auc_score(y, y_score_cv))
            
            pooled_y.append(y)
            pooled_score.append(y_score_cv)
            
            # Fit final RF on full data for feature_importances_
            rf_full = RandomForestClassifier(
                n_estimators=n_estimators,
                max_depth=max_depth,
                min_samples_leaf=min_samples_leaf,
                random_state=random_state,
                n_jobs=n_jobs
            )
            rf_full.fit(X, y)
            importances = pd.Series(rf_full.feature_importances_, index=features).sort_values(ascending=False)
            
            per_model_rows.append({
                "Model": model_name,
                "N": int(len(m_df)),
                "CV_folds_used": int(n_splits),
                "AUROC_CV": auroc_cv
            })
            
            for feat, imp in importances.items():
                importances_rows.append({
                    "Model": model_name,
                    "feature": feat,
                    "importance": float(imp)
                })
        
        per_model_df = pd.DataFrame(per_model_rows).sort_values("Model").reset_index(drop=True)
        importances_df = pd.DataFrame(importances_rows)
        
        # Overall pooled AUROC
        overall_df = pd.DataFrame([{
            "Strategy": strategy_name,
            "Models_included": int(per_model_df["Model"].nunique()) if not per_model_df.empty else 0,
            "Total_N": int(per_model_df["N"].sum()) if not per_model_df.empty else 0,
            "AUROC_CV_pooled": (
                float(roc_auc_score(np.concatenate(pooled_y), np.concatenate(pooled_score)))
                if len(pooled_y) > 0 else np.nan
            )
        }])
        
        show_df(per_model_df, f"Random Forest: {strategy_name} per-model CV AUROC")
        show_df(overall_df, f"Random Forest: {strategy_name} pooled CV AUROC")
        
        all_results[strategy_name] = {
            "per_model_df": per_model_df,
            "overall_df": overall_df,
            "importances_df": importances_df
        }
    
    return all_results
```

```python
def run_rf_all_strategies_analysis(
    results_df,
    min_obs=10,
    show_tables=True,
    # Random Forest params
    n_estimators=500,
    max_depth=None,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1,
    # CV params
    cv_folds=10
):
    """
    Random Forest classifier for all 4 strategies with strategy-specific features.
    
    Strategy 1: inf_diff_z, bc_diff_z
    Strategy 2: inf_diff_z, bc_diff_z, cap_z
    Strategy 3: inf_diff_z, bc_diff_z, cap_z, income_loss_diff_z
    Strategy 4: inf_diff_z, bc_diff_z, cap_z, income_loss_diff_z, preferred_option_num_z
    
    Outputs:
      - all_results: dict with per_model_df, overall_df, importances_df for each strategy
      
    Note:
      - importances_df now contains permutation importance values rather than
        RandomForest feature_importances_ values.
      - Output structure/column names are preserved.
    """
    import pandas as pd
    import numpy as np
    from sklearn.ensemble import RandomForestClassifier
    from sklearn.metrics import roc_auc_score
    from sklearn.model_selection import StratifiedKFold
    from sklearn.inspection import permutation_importance
    
    
    # --- Helper: NumPy z-score ---
    def np_zscore(series: pd.Series) -> pd.Series:
        arr = np.asarray(series, dtype=float)
        mu = np.nanmean(arr)
        sigma = np.nanstd(arr, ddof=0)
        if (not np.isfinite(sigma)) or sigma == 0:
            return pd.Series(np.full_like(arr, np.nan, dtype=float), index=series.index)
        return pd.Series((arr - mu) / sigma, index=series.index)
    
    
    # --- Display helper ---
    def show_df(df, title):
        if not show_tables:
            return
        print("\n" + "=" * 60)
        print(title)
        print("=" * 60)
        try:
            from IPython.display import display
            display(df)
        except Exception:
            print(df.to_string(index=False))
    
    
    # --- Strategy-specific feature definitions (in order) ---
    strategy_features = {
        "strategy_one": ["inf_diff_z", "bc_diff_z"],
        "strategy_two": ["inf_diff_z", "bc_diff_z", "cap_z"],
        "strategy_three": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z"],
        "strategy_four": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z", "preferred_option_num_z"]
    }
    
    
    def add_features(m_df: pd.DataFrame, features: list) -> pd.DataFrame:
        m_df = m_df.copy()
        
        # Z-score all potential features
        m_df["inf_diff_z"] = np_zscore(m_df["inf_diff"])
        m_df["bc_diff_z"]  = np_zscore(m_df["bc_diff"])
        
        if "cap_z" in features:
            m_df["cap_z"] = np_zscore(m_df["capacity_context"])
        
        if "income_loss_diff_z" in features:
            # Exogenous income loss differential: option_1 minus option_2, independent of choice
            m_df["income_loss_1"] = np.where(m_df["choice"] == 1, m_df["income_loss_selected"], m_df["income_loss_non_selected"])
            m_df["income_loss_2"] = np.where(m_df["choice"] == 1, m_df["income_loss_non_selected"], m_df["income_loss_selected"])
            m_df["income_loss_diff"] = m_df["income_loss_1"] - m_df["income_loss_2"]
            m_df["income_loss_diff_z"] = np_zscore(m_df["income_loss_diff"])
        
        if "preferred_option_num_z" in features:
            m_df["preferred_option_num_z"] = np_zscore(m_df["preferred_option_num"])
        
        return m_df.dropna(subset=["choice"] + features)
    
    
    # --- Process each strategy ---
    all_results = {}
    
    for strategy_name, features in strategy_features.items():
        print(f"\n{'='*60}")
        print(f"Processing {strategy_name}")
        print(f"{'='*60}")
        
        # Filter strategy
        df = results_df[results_df["strategy"] == strategy_name].copy()
        print(f"Found {len(df)} rows for {strategy_name}")
        
        if df.empty:
            all_results[strategy_name] = {
                "per_model_df": pd.DataFrame(),
                "overall_df": pd.DataFrame(),
                "importances_df": pd.DataFrame()
            }
            continue
        
        # Ensure numeric cols
        num_cols = [
            "infection_rate_selected", "infection_rate_non_selected",
            "npi_level_selected", "npi_level_non_selected"
        ]
        
        if "cap_z" in features:
            num_cols.append("capacity_context")
        if "income_loss_diff_z" in features:
            num_cols.append("income_loss_selected")
            num_cols.append("income_loss_non_selected")
        if "preferred_option_num_z" in features:
            num_cols.append("preferred_option_num")
        
        for c in num_cols:
            if c in df.columns:
                df[c] = pd.to_numeric(df[c], errors="coerce")
        
        # Choice & attribute reconstruction
        df["choice"] = (df["selected_choice"] == df["pair"].str.split("-").str[0]).astype(int)
        df["inf_1"] = np.where(df["choice"] == 1, df["infection_rate_selected"], df["infection_rate_non_selected"])
        df["inf_2"] = np.where(df["choice"] == 1, df["infection_rate_non_selected"], df["infection_rate_selected"])
        df["npi_1"] = np.where(df["choice"] == 1, df["npi_level_selected"], df["npi_level_non_selected"])
        df["npi_2"] = np.where(df["choice"] == 1, df["npi_level_non_selected"], df["npi_level_selected"])
        
        df["bc_diff"] = df["npi_1"] - df["npi_2"]
        df["inf_diff"] = df["inf_1"] - df["inf_2"]
        
        # Fit per model
        per_model_rows = []
        importances_rows = []
        pooled_y = []
        pooled_score = []
        
        models_sorted = sorted(df["model"].dropna().unique())
        
        for model_name in models_sorted:
            m_df = df[df["model"] == model_name].copy()
            
            # Build required columns list for dropna
            required_cols = ["choice", "bc_diff", "inf_diff"]
            if "cap_z" in features:
                required_cols.append("capacity_context")
            if "income_loss_diff_z" in features:
                required_cols.append("income_loss_selected")
                required_cols.append("income_loss_non_selected")
            if "preferred_option_num_z" in features:
                required_cols.append("preferred_option_num")
            
            m_df = m_df.dropna(subset=required_cols)
            
            if len(m_df) < min_obs:
                continue
            
            m_df = add_features(m_df, features)
            if len(m_df) < min_obs:
                continue
            
            X = m_df[features].to_numpy(dtype=float)
            y = m_df["choice"].to_numpy(dtype=int)
            
            # Must have both classes
            vals, counts = np.unique(y, return_counts=True)
            if len(vals) < 2:
                continue
            
            # Folds can't exceed minority class count
            min_class = int(np.min(counts))
            n_splits = int(min(cv_folds, min_class))
            if n_splits < 2:
                continue
            
            # CV predicted probabilities for AUROC
            cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=random_state)
            y_score_cv = np.full_like(y, fill_value=np.nan, dtype=float)
            
            for fold_i, (tr, te) in enumerate(cv.split(X, y)):
                rf = RandomForestClassifier(
                    n_estimators=n_estimators,
                    max_depth=max_depth,
                    min_samples_leaf=min_samples_leaf,
                    random_state=random_state + fold_i,
                    n_jobs=n_jobs
                )
                rf.fit(X[tr], y[tr])
                y_score_cv[te] = rf.predict_proba(X[te])[:, 1]
            
            auroc_cv = float(roc_auc_score(y, y_score_cv))
            
            pooled_y.append(y)
            pooled_score.append(y_score_cv)
            
            # Fit final RF on full data for permutation importance
            rf_full = RandomForestClassifier(
                n_estimators=n_estimators,
                max_depth=max_depth,
                min_samples_leaf=min_samples_leaf,
                random_state=random_state,
                n_jobs=n_jobs
            )
            rf_full.fit(X, y)
            
            # Permutation importance on the same full dataset, using AUROC scoring
            perm = permutation_importance(
                rf_full,
                X,
                y,
                scoring="roc_auc",
                n_repeats=30,
                random_state=random_state,
                n_jobs=n_jobs
            )
            
            importances = pd.Series(
                perm.importances_mean,
                index=features
            ).sort_values(ascending=False)
            
            per_model_rows.append({
                "Model": model_name,
                "N": int(len(m_df)),
                "CV_folds_used": int(n_splits),
                "AUROC_CV": auroc_cv
            })
            
            for feat, imp in importances.items():
                importances_rows.append({
                    "Model": model_name,
                    "feature": feat,
                    "importance": float(imp)
                })
        
        per_model_df = pd.DataFrame(per_model_rows).sort_values("Model").reset_index(drop=True)
        importances_df = pd.DataFrame(importances_rows)
        
        # Overall pooled AUROC
        overall_df = pd.DataFrame([{
            "Strategy": strategy_name,
            "Models_included": int(per_model_df["Model"].nunique()) if not per_model_df.empty else 0,
            "Total_N": int(per_model_df["N"].sum()) if not per_model_df.empty else 0,
            "AUROC_CV_pooled": (
                float(roc_auc_score(np.concatenate(pooled_y), np.concatenate(pooled_score)))
                if len(pooled_y) > 0 else np.nan
            )
        }])
        
        show_df(per_model_df, f"Random Forest: {strategy_name} per-model CV AUROC")
        show_df(overall_df, f"Random Forest: {strategy_name} pooled CV AUROC")
        
        all_results[strategy_name] = {
            "per_model_df": per_model_df,
            "overall_df": overall_df,
            "importances_df": importances_df
        }
    
    return all_results
```

```python
def visualize_rf_importances(
    all_results,
    # Visualization params
    row_height=5.5,
    bar_height=0.25,
    column_spacing=0.5,
    feature_label_mapping=None,
    legend_position = 1.38
):
    """
    Visualize Random Forest feature importances for all strategies.
    
    Parameters:
    -----------
    all_results : dict
        Output from run_rf_all_strategies_analysis()
    row_height : float
        Height of the figure in inches (default=5.5)
    bar_height : float
        Thickness of individual bars (default=0.25)
    column_spacing : float
        Horizontal spacing between panels (default=0.5)
    feature_label_mapping : dict, optional
        Dictionary to map feature names to display labels
        Example: {'income_loss_diff_z': 'Income Loss Differential', 'cap_z': 'Capacity Context'}
    """
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import matplotlib.patches as mpatches
    import seaborn as sns
    
    
    # --- Styling Configuration ---
    MODEL_NAME_MAPPING = {
        'x-ai/grok-4-fast': 'Grok',
        'openai/gpt-5-nano': 'ChatGPT',
        'google/gemini-2.5-flash': 'Gemini',
        'deepseek/deepseek-chat-v3.1': 'DeepSeek'
    }
    
    MODEL_ORDER = ['ChatGPT', 'DeepSeek', 'Gemini', 'Grok']
    
    _palette_raw = sns.color_palette("tab10", n_colors=len(MODEL_ORDER))
    MODEL_COLORS = dict(zip(MODEL_ORDER, _palette_raw))
    
    BASE_FONT_SIZE = 18
    DPI = 600
    YLABEL_PADDING = 15
    
    # Default feature label mapping if not provided
    if feature_label_mapping is None:
        feature_label_mapping = {}
    
    sns.set_theme(style="whitegrid", context="talk")
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']
    plt.rcParams['figure.dpi'] = DPI
    
    
    # --- Strategy titles ---
    strategy_titles = {
        "strategy_one": "Prompting Strategy One",
        "strategy_two": "Prompting Strategy Two",
        "strategy_three": "Prompting Strategy Three",
        "strategy_four": "Prompting Strategy Four"
    }
    
    strategy_features = {
        "strategy_one": ["inf_diff_z", "bc_diff_z"],
        "strategy_two": ["inf_diff_z", "bc_diff_z", "cap_z"],
        "strategy_three": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z"],
        "strategy_four": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z", "preferred_option_num_z"]
    }
    
    
    # --- Visualization: Single row of 4 panels ---
    n_cols = 4
    fig = plt.figure(figsize=(25, row_height))
    gs = fig.add_gridspec(
        1, n_cols,
        hspace=1,
        wspace=column_spacing
    )
    
    axes = [fig.add_subplot(gs[0, j]) for j in range(n_cols)]
    
    # Create legend handles
    legend_handles = [
        mpatches.Patch(facecolor=MODEL_COLORS[model], edgecolor="black", alpha=0.85, label=model)
        for model in MODEL_ORDER
    ]
    
    # Create stable color palette
    palette = MODEL_COLORS
    
    for idx, (strategy_name, features) in enumerate(strategy_features.items()):
        ax = axes[idx]
        importances_df = all_results[strategy_name]["importances_df"]
        
        if importances_df.empty:
            ax.text(0.5, 0.5, 'No Data', ha='center', va='center', fontsize=BASE_FONT_SIZE)
            ax.set_title(
                strategy_titles[strategy_name],
                fontweight='bold',
                pad=50,
                fontsize=BASE_FONT_SIZE * 1.3
            )
            ax.axis('off')
            continue
        
        # Map model names to display names
        importances_df = importances_df.copy()
        importances_df['Model_Display'] = importances_df['Model'].map(MODEL_NAME_MAPPING)
        importances_df['Model_Display'] = importances_df['Model_Display'].fillna(importances_df['Model'])
        
        # Pivot for grouped bar chart
        pivot_df = importances_df.pivot(index='feature', columns='Model_Display', values='importance')
        
        # Reorder columns (models) alphabetically
        available_models = [m for m in MODEL_ORDER if m in pivot_df.columns]
        pivot_df = pivot_df[available_models]
        
        # Reorder rows (features) in the order they appear in the feature list
        available_features = [f for f in features if f in pivot_df.index]
        pivot_df = pivot_df.reindex(available_features)
        
        # Set bar width and positions
        n_features = len(pivot_df)
        n_models = len(pivot_df.columns)
        y_positions = np.arange(n_features)
        
        # Plot bars for each model
        for i, model in enumerate(pivot_df.columns):
            offset = (i - n_models/2 + 0.5) * bar_height
            color = palette.get(model, 'gray')
            ax.barh(
                y_positions + offset,
                pivot_df[model],
                height=bar_height,
                label=model,
                color=color,
                alpha=0.85,
                edgecolor='black',
                linewidth=1.2
            )
        
        # Customize plot
        ax.set_title(
            strategy_titles[strategy_name],
            fontweight='bold',
            pad=15,
            fontsize=BASE_FONT_SIZE * 1.3
        )
        
        ax.set_xlabel(
            'Feature Importance',
            fontweight='bold',
            fontsize=BASE_FONT_SIZE * 1.0
        )
        
        # Y-label only on leftmost panel
        if idx == 0:
            ax.set_ylabel(
                'Feature',
                fontweight='bold',
                fontsize=BASE_FONT_SIZE * 1.1,
                labelpad=YLABEL_PADDING
            )
        else:
            ax.set_ylabel('')
        
        # Show y-tick labels in all panels with custom mapping
        ax.set_yticks(y_positions)
        feature_labels = [feature_label_mapping.get(f, f) for f in pivot_df.index]
        ax.set_yticklabels(feature_labels, fontsize=BASE_FONT_SIZE * 0.9)
        
        # Grid - only vertical (x-axis)
        ax.grid(True, axis='x', alpha=0.25, linestyle='--')
        ax.grid(False, axis='y')  # Explicitly disable horizontal gridlines
        ax.set_axisbelow(True)
        ax.tick_params(axis='both', labelsize=BASE_FONT_SIZE * 0.9)
    
    # Add legend at top
    fig.legend(
        handles=legend_handles,
        title="Model",
        loc="upper center",
        bbox_to_anchor=(0.5, legend_position),
        ncol=len(MODEL_ORDER),
        fontsize=BASE_FONT_SIZE * 0.95,
        title_fontsize=BASE_FONT_SIZE * 1.05,
        frameon=True,
        edgecolor="black",
        fancybox=False
    )
    
    sns.despine()
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.show()
```

```python
# --- Reasoning Component Visualization Configuration ---
# Edit this dictionary to change labels or include/exclude components in plots.
# The order of keys here will determine the order in the visualizations.
REASONING_COMPONENT_CONFIG = {
    "mentions_hospital_capacity": "Mentions Hospital Occupancy",
    "computed_hospital_demand": "Computed Hospital Demand",
    "compares_demand_to_capacity": "Compares Hospitalizations to Occupancy",
    "short_term_economic_reasoning": "Short Term Economic Reasoning",
    "long_term_economic_reasoning": "Long Term Economic Reasoning",
    "mentions_business_closures": "Mentions Business Closures",
    "compares_infection_rates_between_options": "Compares Infection Risk",
    "argues_closures_not_worth_it_given_infections": "Discusses Closure-Infection Tradeoff",
    "mentions_user_preference": "Mentions User Preference",
    "mentions_vulnerable_groups": "Mentions Vulnerable Groups",
    "acknowledges_data_uncertainty": "Acknowledges Data Uncertainty",
    "expresses_recommendation_caution": "Recommends with Caution",
    "explicitly_states_tradeoff": "Explicitly States Tradeoff"
}

def visualize_reasoning_components(
    reasoning_df,
    MODEL_COLORS = MODEL_COLORS,
    components_to_display=None,
    # Visualization params
    row_height=5.5,
    column_spacing=0.1,
    y_label_wrap_width=20,
    y_label_padding=1,
    bar_group_width=0.8,
    legend_position = 1.38
):
    """
    Visualize reasoning components across strategies in a single row.
    
    Parameters:
    -----------
    reasoning_df : DataFrame
        DataFrame containing reasoning component data
    MODEL_COLORS : dict
        Dictionary mapping model names to colors
    components_to_display : list, optional
        List of component keys from REASONING_COMPONENT_CONFIG to display.
        If None, uses all components. Order matters.
        Example: ['short_term_economic_reasoning', 'mentions_hospital_capacity',
                  'long_term_economic_reasoning', 'mentions_user_preference']
    row_height : float
        Height of the figure in inches (default=5.5)
    column_spacing : float
        Horizontal spacing between panels (default=0.5)
    y_label_wrap_width : int
        Character width for wrapping y-axis labels (default=20)
    y_label_padding : float
        Spacing between y-axis labels and bars (default=1)
    bar_group_width : float
        Width of bar groups, controls vertical spacing (default=0.8)
    """
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import matplotlib.patches as mpatches
    import seaborn as sns
    import string
    import textwrap
    
    
    # --- Styling Configuration ---
    BASE_FONT_SIZE = 18
    DPI = 600
    
    sns.set_theme(style="whitegrid", context="talk")
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']
    plt.rcParams['figure.dpi'] = DPI
    
    
    # --- Helper functions ---
    def extract_strategy_number(strategy_name):
        """Extract numeric part from strategy name for sorting"""
        import re
        match = re.search(r'\d+', strategy_name)
        return int(match.group()) if match else 0
    
    
    def format_strategy_name(strategy):
        """Format strategy name for display"""
        if isinstance(strategy, str):
            if strategy.startswith('strategy_'):
                num = strategy.split('_')[1].title()
                return f"Strategy {num}"
        return str(strategy).title()
    
    
    # --- Determine which components to display ---
    if components_to_display is None:
        # Use all components from config
        components_to_display = list(REASONING_COMPONENT_CONFIG.keys())
    
    # Filter config to only selected components (maintaining order)
    display_config = {k: REASONING_COMPONENT_CONFIG[k] 
                     for k in components_to_display 
                     if k in REASONING_COMPONENT_CONFIG}
    
    if len(display_config) == 0:
        print("No valid components to display")
        return
    
    
    # --- Prepare data ---
    if reasoning_df.empty:
        print("Reasoning dataframe is empty")
        return
    
    # Melt dataframe for selected components only
    df_reasoning = reasoning_df.melt(
        id_vars=['model', 'strategy', 'choice_status'], 
        value_vars=list(display_config.keys()), 
        var_name='component', 
        value_name='present'
    )
    
    # Map component names to labels
    df_reasoning['component_label'] = df_reasoning['component'].map(display_config)
    
    # Ensure component_label order matches display_config order
    component_label_order = list(display_config.values())
    df_reasoning['component_label'] = pd.Categorical(
        df_reasoning['component_label'],
        categories=component_label_order,
        ordered=True
    )
    
    
    # --- Get sorted strategies ---
    strategies_sorted = sorted(reasoning_df['strategy'].unique(), key=extract_strategy_number)
    n_strategies = len(strategies_sorted)
    
    strategy_titles = {
        "strategy_one": "Prompting Strategy One",
        "strategy_two": "Prompting Strategy Two",
        "strategy_three": "Prompting Strategy Three",
        "strategy_four": "Prompting Strategy Four"
    }
    
    
    # --- Visualization: Single row ---
    n_cols = 4  # Always 4 columns for 4 strategies
    fig = plt.figure(figsize=(25, row_height))
    gs = fig.add_gridspec(
        1, n_cols,
        hspace=1,
        wspace=column_spacing
    )
    
    axes = [fig.add_subplot(gs[0, j]) for j in range(n_cols)]
    
    # Get sorted models for consistent legend
    sorted_models = sorted(reasoning_df['model'].unique())
    
    for idx, strategy in enumerate(strategies_sorted):
        if idx >= n_cols:
            break
            
        ax = axes[idx]
        
        # Filter data for this strategy
        strategy_data = df_reasoning[df_reasoning['strategy'] == strategy].copy()
        
        # Sort by component_label to maintain order
        strategy_data = strategy_data.sort_values('component_label')
        
        sns.barplot(
            data=strategy_data,
            x='present', 
            y='component_label', 
            hue='model',
            hue_order=sorted_models,
            palette=MODEL_COLORS,
            errorbar=None,
            ax=ax,
            width=bar_group_width,
            orient='h',
            edgecolor='black',  # Add black border
            linewidth=1.2       # Border thickness
        )
        
        # Set panel title
        ax.set_title(
            strategy_titles.get(strategy, format_strategy_name(strategy)),
            fontsize=BASE_FONT_SIZE * 1.3,
            fontweight='bold',
            pad=15
        )
        
        # Set axis labels
        ax.set_xlabel('Frequency', fontsize=BASE_FONT_SIZE * 1.0, fontweight='bold')
        
        # Y-label only on leftmost panel
        if idx == 0:
            ax.set_ylabel('Reasoning Component', fontsize=BASE_FONT_SIZE * 1.1, fontweight='bold', labelpad=15)
        else:
            ax.set_ylabel('')
        
        # Wrap y-axis labels
        y_labels = [label.get_text() for label in ax.get_yticklabels()]
        wrapped_labels = [textwrap.fill(label, width=y_label_wrap_width) for label in y_labels]
        ax.set_yticklabels(wrapped_labels if idx == 0 else [])
        
        # Adjust tick sizes
        ax.tick_params(axis='y', labelsize=BASE_FONT_SIZE * 0.9, pad=y_label_padding)
        ax.tick_params(axis='x', labelsize=BASE_FONT_SIZE * 0.9)
        
        # Remove individual legends from each panel
        if ax.get_legend() is not None:
            ax.get_legend().remove()
        
        # Grid - only vertical
        ax.grid(True, axis='x', alpha=0.25, linestyle='--')
        ax.grid(False, axis='y')  # Disable horizontal gridlines
        ax.set_axisbelow(True)
    
    # Hide unused panels if there are fewer than 4 strategies
    for idx in range(n_strategies, n_cols):
        axes[idx].set_visible(False)
    
    # Create model legend handles
    model_handles = [
        mpatches.Patch(facecolor=MODEL_COLORS[m], edgecolor='black', label=m, alpha=0.85)
        for m in sorted_models
    ]
    
    # Add legend at top
    fig.legend(
        handles=model_handles, 
        #title='Model', 
        loc='upper center',
        bbox_to_anchor=(0.5, legend_position),
        ncol=len(sorted_models),
        fontsize=BASE_FONT_SIZE * 0.95,
        title_fontsize=BASE_FONT_SIZE * 1.05,
        frameon=True,
        edgecolor='black',
        fancybox=False
    )
    
    sns.despine()
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.show()
```

```python
# --- Reasoning Component Visualization Configuration ---
# Edit this dictionary to change labels or include/exclude components in plots.
# The order of keys here will determine the order in the visualizations.
REASONING_COMPONENT_CONFIG = {
    "mentions_hospital_capacity": "Mentions Hospital Capacity",
    "computed_hospital_demand": "Computed Hospital Demand",
    "compares_demand_to_capacity": "Compares Demand to Capacity",
    "short_term_economic_reasoning": "Short Term Economic Reasoning",
    "long_term_economic_reasoning": "Long Term Economic Reasoning",
    "mentions_business_closures": "Mentions Business Closures",
    "compares_infection_rates_between_options": "Compares Infection Risk",
    "argues_closures_not_worth_it_given_infections": "Discusses Closure-Infection Tradeoff",
    "mentions_user_preference": "Mentions User Preference",
    "mentions_vulnerable_groups": "Mentions Vulnerable Groups",
    "acknowledges_data_uncertainty": "Acknowledges Data Uncertainty",
    "expresses_recommendation_caution": "Recommends with Caution",
    "explicitly_states_tradeoff": "Explicitly States Tradeoff"
}


def summarize_reasoning_components(
    reasoning_df,
    components_to_display=None,
    display_tables=True
):
    """
    Generate frequency tables for reasoning components across strategies and models.
    
    Parameters:
    -----------
    reasoning_df : DataFrame
        DataFrame containing reasoning component data
    components_to_display : list, optional
        List of component keys from REASONING_COMPONENT_CONFIG to display.
        If None, uses all components. Order matters.
        Example: ['short_term_economic_reasoning', 'mentions_hospital_capacity',
                  'long_term_economic_reasoning', 'mentions_user_preference']
    display_tables : bool
        Whether to print tables to console (default=True)
    
    Returns:
    --------
    dict : Dictionary containing frequency tables for each strategy
        Keys are strategy names, values are DataFrames with frequencies
    """
    import pandas as pd
    import numpy as np
    
    # --- Helper functions ---
    def extract_strategy_number(strategy_name):
        """Extract numeric part from strategy name for sorting"""
        import re
        match = re.search(r'\d+', strategy_name)
        return int(match.group()) if match else 0
    
    
    def format_strategy_name(strategy):
        """Format strategy name for display"""
        if isinstance(strategy, str):
            if strategy.startswith('strategy_'):
                num = strategy.split('_')[1].title()
                return f"Strategy {num}"
        return str(strategy).title()
    
    
    # --- Determine which components to display ---
    if components_to_display is None:
        # Use all components from config
        components_to_display = list(REASONING_COMPONENT_CONFIG.keys())
    
    # Filter config to only selected components (maintaining order)
    display_config = {k: REASONING_COMPONENT_CONFIG[k] 
                     for k in components_to_display 
                     if k in REASONING_COMPONENT_CONFIG}
    
    if len(display_config) == 0:
        print("No valid components to display")
        return {}
    
    
    # --- Prepare data ---
    if reasoning_df.empty:
        print("Reasoning dataframe is empty")
        return {}
    
    # Get sorted strategies
    strategies_sorted = sorted(reasoning_df['strategy'].unique(), key=extract_strategy_number)
    sorted_models = sorted(reasoning_df['model'].unique())
    
    # Create results dictionary
    results = {}
    
    # Process each strategy
    for strategy in strategies_sorted:
        strategy_data = reasoning_df[reasoning_df['strategy'] == strategy].copy()
        
        # Calculate frequencies for each component and model
        freq_rows = []
        
        for component_key in components_to_display:
            if component_key not in strategy_data.columns:
                continue
                
            component_label = display_config[component_key]
            
            # Calculate frequency for each model
            row_data = {'Component': component_label}
            
            for model in sorted_models:
                model_data = strategy_data[strategy_data['model'] == model]
                if len(model_data) > 0:
                    frequency = model_data[component_key].mean()
                    count = model_data[component_key].sum()
                    total = len(model_data)
                    row_data[model] = frequency
                    row_data[f'{model}_count'] = f"{int(count)}/{total}"
                else:
                    row_data[model] = np.nan
                    row_data[f'{model}_count'] = "0/0"
            
            freq_rows.append(row_data)
        
        # Create DataFrame
        freq_df = pd.DataFrame(freq_rows)
        
        # Create separate tables for frequencies and counts
        freq_columns = ['Component'] + sorted_models
        count_columns = ['Component'] + [f'{m}_count' for m in sorted_models]
        
        freq_table = freq_df[freq_columns].copy()
        count_table = freq_df[count_columns].copy()
        
        # Rename count columns for clarity
        count_table.columns = ['Component'] + sorted_models
        
        results[strategy] = {
            'frequencies': freq_table,
            'counts': count_table,
            'combined': freq_df
        }
        
        # Display tables if requested
        if display_tables:
            print("\n" + "="*80)
            print(f"{format_strategy_name(strategy)} - Reasoning Component Frequencies")
            print("="*80)
            print("\nFrequencies (proportion of responses with component):")
            try:
                from IPython.display import display
                display(freq_table.round(3))
            except:
                print(freq_table.round(3).to_string(index=False))
            
            print("\nCounts (present/total):")
            try:
                from IPython.display import display
                display(count_table)
            except:
                print(count_table.to_string(index=False))
    
    return results


# Example usage:
if not reasoning_df.empty:
    # Option 1: Specific components
    selected_components = [
        'short_term_economic_reasoning',
        'mentions_hospital_capacity',
        'long_term_economic_reasoning',
        'mentions_user_preference'
    ]
    
    tables = summarize_reasoning_components(
        reasoning_df,
        components_to_display=selected_components,
        display_tables=True
    )
    
    # Access individual tables
    # strategy_1_freq = tables['strategy_one']['frequencies']
    # strategy_1_counts = tables['strategy_one']['counts']
    
    # Option 2: All components
    # all_tables = summarize_reasoning_components(reasoning_df)
```

### Analysis Function

```python
def run_elasticnet_all_strategies_analysis(
    results_df,
    min_obs=10,
    show_tables=True,
    # Joint CV params
    cv_folds=10,
    # Elastic Net grid
    l1_ratios=np.linspace(0.1, 1.0, 10),
    alphas=np.logspace(-4, 1, 20),
    random_state=42,
    n_jobs=-1,
    # Bootstrap params for CIs (final refit only)
    n_bootstrap=1000,
    ci_percentile=95,
    # Feature selection params (within-fold PI)
    perm_importance_threshold=0.05,
    n_perm_repeats=5,
    # How to aggregate fold-wise selected features into one final set
    final_feature_rule="majority",   # {"majority","union","intersection"}
    final_feature_majority_frac=0.5, # used if final_feature_rule=="majority"
):
    """
    Elastic-Net Logistic Regression analysis for all 4 strategies with feature interactions.
    Computed separately per LLM model with model-specific JOINT feature selection + hyperparameter tuning.

    For each strategy and each LLM:
      - For each (alpha, l1_ratio) combination:
        - For each CV fold:
          1) Standardize using training fold only
          2) Fit elastic-net on training fold
          3) Compute permutation importance on training fold using AUROC (scoring='roc_auc')
          4) Keep features with importance >= perm_importance_threshold
          5) Refit elastic-net on training fold with kept features
          6) Evaluate AUROC on held-out fold
      - Choose the (alpha, l1_ratio) combo with best mean fold AUROC
      - Aggregate fold-selected features (majority/union/intersection) for that best combo
      - Refit final elastic-net on all data using those features
      - Bootstrap coefficients (on all-data refit) to get CIs

    Notes:
      - Permutation importance uses the scoring function provided; here we use AUROC via scoring='roc_auc'. [web:21][web:2]
      - AUROC in scikit-learn is computed from probability estimates when using predict_proba. [web:51]
      - StratifiedKFold preserves class proportions across folds. [web:61]

    Returns:
      - all_results: dict with results for each strategy
      - auroc_summary: DataFrame with AUROC summary table
    """
    import pandas as pd
    import numpy as np
    from sklearn.linear_model import LogisticRegression
    from sklearn.preprocessing import StandardScaler
    from sklearn.model_selection import StratifiedKFold
    from sklearn.inspection import permutation_importance
    from sklearn.metrics import roc_auc_score
    from sklearn.utils import resample
    from itertools import combinations
    import warnings
    warnings.filterwarnings("ignore")

    # --- Model mappings ---
    MODEL_NAME_MAPPING = {
        "x-ai/grok-4-fast": "Grok",
        "openai/gpt-5-nano": "ChatGPT",
        "google/gemini-2.5-flash": "Gemini",
        "deepseek/deepseek-chat-v3.1": "DeepSeek",
    }
    MODEL_ORDER = ["ChatGPT", "DeepSeek", "Gemini", "Grok"]

    # --- Strategy definitions ---
    strategy_base_features = {
        "strategy_one": ["inf_diff_z", "bc_diff_z"],
        "strategy_two": ["inf_diff_z", "bc_diff_z", "cap_z"],
        "strategy_three": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z"],
        "strategy_four": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z", "preferred_option_num_z"],
    }
    strategy_titles = {
        "strategy_one": "Strategy 1",
        "strategy_two": "Strategy 2",
        "strategy_three": "Strategy 3",
        "strategy_four": "Strategy 4",
    }

    # --- Helper: NumPy z-score ---
    def np_zscore(series: pd.Series) -> pd.Series:
        arr = np.asarray(series, dtype=float)
        mu = np.nanmean(arr)
        sigma = np.nanstd(arr, ddof=0)
        if (not np.isfinite(sigma)) or sigma == 0:
            return pd.Series(np.full_like(arr, np.nan, dtype=float), index=series.index)
        return pd.Series((arr - mu) / sigma, index=series.index)

    # --- Display helper ---
    def show_df(df, title):
        if not show_tables:
            return
        print("\n" + "=" * 80)
        print(title)
        print("=" * 80)
        try:
            from IPython.display import display
            display(df)
        except Exception:
            print(df.to_string(index=False))

    # --- Create polynomial interaction features ---
    def create_interaction_features(df, base_features):
        """
        Create interaction features without repeated terms.
        Excludes:
        - cap_z as main effect (option-independent context variable)
        - Any purely option-independent combinations
        Includes:
        - All interactions with at least one option-dependent feature
          (inf_diff_z, bc_diff_z, income_loss_diff_z, preferred_option_num_z)
        """
        interaction_df = df[base_features].copy()
        feature_names = list(base_features)

        option_independent = {"cap_z"}
        option_dependent = {"inf_diff_z", "bc_diff_z", "income_loss_diff_z", "preferred_option_num_z"}

        max_degree = len(base_features)
        for degree in range(2, max_degree + 1):
            for combo in combinations(base_features, degree):
                combo_set = set(combo)

                if combo_set.issubset(option_independent):
                    continue
                if not combo_set.intersection(option_dependent):
                    continue

                feature_name = "_x_".join(combo)
                interaction_df[feature_name] = df[combo[0]]
                for feat in combo[1:]:
                    interaction_df[feature_name] *= df[feat]
                feature_names.append(feature_name)

        return interaction_df, feature_names

    # --- Add base features ---
    def add_base_features(m_df: pd.DataFrame, strategy_name: str) -> pd.DataFrame:
        m_df = m_df.copy()
        m_df["inf_diff_z"] = np_zscore(m_df["inf_diff"])
        m_df["bc_diff_z"] = np_zscore(m_df["bc_diff"])

        if strategy_name in ["strategy_two", "strategy_three", "strategy_four"]:
            m_df["cap_z"] = np_zscore(m_df["capacity_context"])
        if strategy_name in ["strategy_three", "strategy_four"]:
            # Exogenous income loss differential: option_1 minus option_2, independent of choice
            m_df["income_loss_1"] = np.where(m_df["choice"] == 1, m_df["income_loss_selected"], m_df["income_loss_non_selected"])
            m_df["income_loss_2"] = np.where(m_df["choice"] == 1, m_df["income_loss_non_selected"], m_df["income_loss_selected"])
            m_df["income_loss_diff"] = m_df["income_loss_1"] - m_df["income_loss_2"]
            m_df["income_loss_diff_z"] = np_zscore(m_df["income_loss_diff"])
        if strategy_name == "strategy_four":
            m_df["preferred_option_num_z"] = np_zscore(m_df["preferred_option_num"])
        return m_df

    # --- Bootstrap CI calculation (final refit only) ---
    def bootstrap_coefficients(X, y, C_opt, l1_ratio_opt, n_bootstrap, random_state):
        np.random.seed(random_state)
        coef_samples = []
        for i in range(n_bootstrap):
            X_boot, y_boot = resample(X, y, random_state=random_state + i)
            if len(np.unique(y_boot)) < 2:
                continue
            model = LogisticRegression(
                penalty="elasticnet",
                solver="saga",
                C=C_opt,
                l1_ratio=l1_ratio_opt,
                max_iter=10000,
                random_state=random_state + i,
            )
            try:
                model.fit(X_boot, y_boot)
                coef_samples.append(model.coef_[0])
            except Exception:
                continue
        return np.array(coef_samples)

    # --- Feature aggregation across folds for best hyperparams ---
    def aggregate_selected_features(selected_masks, feature_names):
        """
        selected_masks: list of boolean arrays (len=p) for each fold (may vary in count if some folds skipped)
        """
        if selected_masks is None or len(selected_masks) == 0:
            return []

        mat = np.vstack(selected_masks).astype(int)  # (n_folds_used, p)
        counts = mat.sum(axis=0)
        n_used = mat.shape[0]

        if final_feature_rule == "union":
            keep = counts >= 1
        elif final_feature_rule == "intersection":
            keep = counts == n_used
        elif final_feature_rule == "majority":
            keep = counts >= int(np.ceil(final_feature_majority_frac * n_used))
        else:
            raise ValueError("final_feature_rule must be one of {'majority','union','intersection'}")

        return [f for f, k in zip(feature_names, keep) if k]

    # --- Joint tuning + within-fold PI selection ---
    def joint_cv_select_and_tune(X, y, feature_names):
        """
        Returns dict with best hyperparams, mean AUROC, per-fold selected masks for best combo,
        and out-of-fold predictions for best combo (for an overall AUROC).
        """
        y = np.asarray(y, dtype=int)
        X = np.asarray(X, dtype=float)

        if len(np.unique(y)) < 2 or len(y) < min_obs:
            return None

        vals, counts = np.unique(y, return_counts=True)
        if len(vals) < 2:
            return None

        n_splits = int(min(cv_folds, int(np.min(counts))))
        if n_splits < 2:
            return None

        cv = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=random_state)

        Cs_grid = 1.0 / np.asarray(alphas, dtype=float)

        best = None
        best_mean_auc = -np.inf

        for C in Cs_grid:
            for l1_ratio in l1_ratios:
                fold_aucs = []
                fold_masks = []
                oof = np.full(len(y), np.nan, dtype=float)

                for fold_i, (tr, te) in enumerate(cv.split(X, y)):
                    X_tr, X_te = X[tr], X[te]
                    y_tr, y_te = y[tr], y[te]

                    # Scale within fold
                    scaler = StandardScaler()
                    X_tr_s = scaler.fit_transform(X_tr)
                    X_te_s = scaler.transform(X_te)

                    # Fit model for permutation importance (same family as final)
                    base = LogisticRegression(
                        penalty="elasticnet",
                        solver="saga",
                        C=float(C),
                        l1_ratio=float(l1_ratio),
                        max_iter=10000,
                        random_state=random_state + fold_i,
                    )
                    try:
                        base.fit(X_tr_s, y_tr)
                    except Exception:
                        continue

                    # Permutation importance on training fold using AUROC
                    perm = permutation_importance(
                        base,
                        X_tr_s,
                        y_tr,
                        scoring="roc_auc",
                        n_repeats=n_perm_repeats,
                        random_state=random_state + fold_i,
                        n_jobs=n_jobs,
                    )
                    keep = perm.importances_mean >= perm_importance_threshold
                    fold_masks.append(keep)

                    if not np.any(keep):
                        continue

                    # Refit and evaluate on holdout fold
                    model = LogisticRegression(
                        penalty="elasticnet",
                        solver="saga",
                        C=float(C),
                        l1_ratio=float(l1_ratio),
                        max_iter=10000,
                        random_state=random_state + 10_000 + fold_i,
                    )
                    try:
                        model.fit(X_tr_s[:, keep], y_tr)
                        proba = model.predict_proba(X_te_s[:, keep])[:, 1]
                    except Exception:
                        continue

                    oof[te] = proba
                    fold_aucs.append(roc_auc_score(y_te, proba))

                if len(fold_aucs) == 0:
                    continue

                mean_auc = float(np.mean(fold_aucs))
                if mean_auc > best_mean_auc:
                    best_mean_auc = mean_auc
                    best = {
                        "C_opt": float(C),
                        "alpha_opt": float(1.0 / C),
                        "l1_ratio_opt": float(l1_ratio),
                        "mean_auc": mean_auc,
                        "n_splits_used": int(n_splits),
                        "fold_masks": fold_masks,
                        "oof_pred": oof,
                    }

        if best is None:
            return None

        # Overall AUROC from OOF predictions (dropping NaNs from skipped folds)
        ok = np.isfinite(best["oof_pred"])
        if ok.sum() > 0 and len(np.unique(y[ok])) == 2:
            best["auroc_oof"] = float(roc_auc_score(y[ok], best["oof_pred"][ok]))
        else:
            best["auroc_oof"] = np.nan

        # Aggregate fold-wise selected features into a final set
        best["selected_features_final"] = aggregate_selected_features(best["fold_masks"], feature_names)

        return best

    # --- Process each strategy ---
    all_results = {}
    auroc_summary_rows = []

    for strategy_name, base_features in strategy_base_features.items():
        print(f"\n{'='*80}")
        print(f"Processing {strategy_name}: {base_features}")
        print(f"{'='*80}")

        df = results_df[results_df["strategy"] == strategy_name].copy()
        print(f"Found {len(df)} rows for {strategy_name}")

        if df.empty:
            all_results[strategy_name] = {
                "per_model_df": pd.DataFrame(),
                "overall_df": pd.DataFrame(),
                "odds_ratios_df": pd.DataFrame(),
            }
            continue

        # Ensure numeric columns
        num_cols = [
            "infection_rate_selected", "infection_rate_non_selected",
            "npi_level_selected", "npi_level_non_selected",
        ]
        if "cap_z" in base_features:
            num_cols.append("capacity_context")
        if "income_loss_diff_z" in base_features:
            num_cols.append("income_loss_selected")
            num_cols.append("income_loss_non_selected")
        if "preferred_option_num_z" in base_features:
            num_cols.append("preferred_option_num")

        for c in num_cols:
            if c in df.columns:
                df[c] = pd.to_numeric(df[c], errors="coerce")

        # Reconstruct choice and differences
        df["choice"] = (df["selected_choice"] == df["pair"].str.split("-").str[0]).astype(int)
        df["inf_1"] = np.where(df["choice"] == 1, df["infection_rate_selected"], df["infection_rate_non_selected"])
        df["inf_2"] = np.where(df["choice"] == 1, df["infection_rate_non_selected"], df["infection_rate_selected"])
        df["npi_1"] = np.where(df["choice"] == 1, df["npi_level_selected"], df["npi_level_non_selected"])
        df["npi_2"] = np.where(df["choice"] == 1, df["npi_level_non_selected"], df["npi_level_selected"])

        df["bc_diff"] = df["npi_1"] - df["npi_2"]
        df["inf_diff"] = df["inf_1"] - df["inf_2"]

        # Add base features
        df = add_base_features(df, strategy_name)

        # Required columns for dropna
        required_cols = ["choice", "inf_diff", "bc_diff"]
        if "cap_z" in base_features:
            required_cols.append("capacity_context")
        if "income_loss_diff_z" in base_features:
            required_cols.append("income_loss_selected")
            required_cols.append("income_loss_non_selected")
        if "preferred_option_num_z" in base_features:
            required_cols.append("preferred_option_num")

        df = df.dropna(subset=required_cols)
        if len(df) < min_obs:
            print(f"Insufficient data after cleaning: {len(df)} rows")
            all_results[strategy_name] = {
                "per_model_df": pd.DataFrame(),
                "overall_df": pd.DataFrame(),
                "odds_ratios_df": pd.DataFrame(),
            }
            continue

        # Create interaction features
        print("Creating interaction features...")
        interaction_df, all_feature_names = create_interaction_features(df, base_features)
        print(f"Created {len(all_feature_names)} features (including interactions)")

        models_sorted = sorted(df["model"].dropna().unique())

        # --- Process Each Model (joint selection + tuning) ---
        per_model_rows = []
        odds_ratios_rows = []
        pooled_y = []
        pooled_score = []

        for model_name in models_sorted:
            m_df = df[df["model"] == model_name].copy()
            if len(m_df) < min_obs:
                continue

            X_full = interaction_df.loc[m_df.index, all_feature_names].to_numpy(dtype=float)
            y = m_df["choice"].to_numpy(dtype=int)

            if len(np.unique(y)) < 2:
                continue

            # Joint selection+tuning using within-fold PI(AUROC)
            best = joint_cv_select_and_tune(X_full, y, all_feature_names)
            if best is None:
                continue

            selected_features = best["selected_features_final"]
            if selected_features is None or len(selected_features) == 0:
                continue

            # Final refit on all data with selected features + best hyperparams
            X = interaction_df.loc[m_df.index, selected_features].to_numpy(dtype=float)

            scaler = StandardScaler()
            X_scaled = scaler.fit_transform(X)

            final_model = LogisticRegression(
                penalty="elasticnet",
                solver="saga",
                C=best["C_opt"],
                l1_ratio=best["l1_ratio_opt"],
                max_iter=10000,
                random_state=random_state,
            )
            try:
                final_model.fit(X_scaled, y)
            except Exception:
                continue

            coefficients = final_model.coef_[0]

            # Record AUROC (use mean fold AUROC from joint CV; also keep OOF AUROC)
            auroc_cv = best["mean_auc"]

            pooled_y.append(y)
            # For pooled AUROC, use OOF if available; else fall back to in-sample probs (not ideal)
            oof = best.get("oof_pred", None)
            if oof is not None and np.isfinite(oof).sum() == len(y):
                pooled_score.append(oof)
            else:
                pooled_score.append(final_model.predict_proba(X_scaled)[:, 1])

            auroc_summary_rows.append({
                "Strategy": strategy_titles[strategy_name],
                "Model": model_name,
                "AUROC_CV": auroc_cv,
                "N": int(len(m_df)),
            })

            # Bootstrap CIs on the final refit design matrix
            coef_samples = bootstrap_coefficients(
                X_scaled, y,
                C_opt=best["C_opt"],
                l1_ratio_opt=best["l1_ratio_opt"],
                n_bootstrap=n_bootstrap,
                random_state=random_state,
            )

            lower_percentile = (100 - ci_percentile) / 2
            upper_percentile = 100 - lower_percentile

            if coef_samples.size == 0:
                coef_lower = np.full_like(coefficients, np.nan, dtype=float)
                coef_upper = np.full_like(coefficients, np.nan, dtype=float)
            else:
                coef_lower = np.percentile(coef_samples, lower_percentile, axis=0)
                coef_upper = np.percentile(coef_samples, upper_percentile, axis=0)

            ors = np.exp(coefficients)
            or_lower = np.exp(coef_lower)
            or_upper = np.exp(coef_upper)

            per_model_rows.append({
                "Model": model_name,
                "N": int(len(m_df)),
                "N_features": int(len(selected_features)),
                "CV_folds_used": int(best["n_splits_used"]),
                "AUROC_CV": float(auroc_cv),
                "AUROC_OOF": float(best.get("auroc_oof", np.nan)),
                "optimal_alpha": float(best["alpha_opt"]),
                "optimal_l1_ratio": float(best["l1_ratio_opt"]),
                "final_feature_rule": str(final_feature_rule),
            })

            for feat, coef, or_val, ci_low, ci_high in zip(
                selected_features, coefficients, ors, or_lower, or_upper
            ):
                odds_ratios_rows.append({
                    "Model": model_name,
                    "feature": feat,
                    "coefficient": float(coef),
                    "odds_ratio": float(or_val),
                    "ci_lower": float(ci_low),
                    "ci_upper": float(ci_high),
                })

        per_model_df = pd.DataFrame(per_model_rows).sort_values("Model").reset_index(drop=True) if len(per_model_rows) else pd.DataFrame()
        odds_ratios_df = pd.DataFrame(odds_ratios_rows) if len(odds_ratios_rows) else pd.DataFrame()

        overall_df = pd.DataFrame([{
            "Strategy": strategy_name,
            "Models_included": int(per_model_df["Model"].nunique()) if not per_model_df.empty else 0,
            "Total_N": int(per_model_df["N"].sum()) if not per_model_df.empty else 0,
            "AUROC_CV_pooled": (
                float(roc_auc_score(np.concatenate(pooled_y), np.concatenate(pooled_score)))
                if len(pooled_y) > 0 else np.nan
            ),
        }])

        show_df(per_model_df, f"Elastic-Net (Joint FS+Tuning): {strategy_name} per-model results")
        if not odds_ratios_df.empty:
            show_df(odds_ratios_df.head(20), f"Elastic-Net (Joint FS+Tuning): {strategy_name} odds ratios (first 20)")

        all_results[strategy_name] = {
            "per_model_df": per_model_df,
            "overall_df": overall_df,
            "odds_ratios_df": odds_ratios_df,
        }

    # --- AUROC Summary Table ---
    auroc_summary_df = pd.DataFrame(auroc_summary_rows)
    auroc_pivot = pd.DataFrame()

    if not auroc_summary_df.empty:
        auroc_pivot = auroc_summary_df.pivot(index="Strategy", columns="Model", values="AUROC_CV")

        auroc_pivot.columns = auroc_pivot.columns.map(lambda x: MODEL_NAME_MAPPING.get(x, x))
        available_models = [m for m in MODEL_ORDER if m in auroc_pivot.columns]
        auroc_pivot = auroc_pivot[available_models]

        strategy_order = ["Strategy 1", "Strategy 2", "Strategy 3", "Strategy 4"]
        auroc_pivot = auroc_pivot.reindex([s for s in strategy_order if s in auroc_pivot.index])

        show_df(auroc_pivot.round(4), "AUROC Summary (Joint FS+Tuning): Cross-Validated AUROC by Strategy and Model")

    return {
        "all_results": all_results,
        "auroc_summary": auroc_pivot,
    }
```

```python
def visualize_elasticnet_forest_plots(
    all_results,
    # Visualization params
    row_height=6.0,
    column_spacing=0.5,
    dot_size=10,
    error_bar_width=2.5,
    feature_label_mapping=None,
    global_xmin=None,
    global_xmax=None
):
    """
    Visualize Elastic-Net odds ratios as forest plots in a single row.

    Parameters:
    -----------
    all_results : dict
        Output from run_elasticnet_all_strategies_analysis()['all_results']
    row_height : float
        Height of the figure in inches (default=6.0)
    column_spacing : float
        Horizontal spacing between panels (default=0.5)
    dot_size : float
        Size of the dots in forest plot (default=10)
    error_bar_width : float
        Width of error bar lines (default=2.5)
    feature_label_mapping : dict, optional
        Dictionary to map feature names to display labels
    global_xmin : float, optional
        Override the minimum x-axis bound across all panels (e.g. 0.3).
        If None, inferred from CI lower bounds in the data.
    global_xmax : float, optional
        Override the maximum x-axis bound across all panels (e.g. 4.0).
        If None, inferred from CI upper bounds in the data.
    """
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import matplotlib.patches as mpatches
    import matplotlib.ticker as mticker
    import seaborn as sns

    # --- Styling Configuration ---
    MODEL_NAME_MAPPING = {
        'x-ai/grok-4-fast': 'Grok',
        'openai/gpt-5-nano': 'ChatGPT',
        'google/gemini-2.5-flash': 'Gemini',
        'deepseek/deepseek-chat-v3.1': 'DeepSeek'
    }
    MODEL_ORDER = ['ChatGPT', 'DeepSeek', 'Gemini', 'Grok']

    _palette_raw = sns.color_palette("tab10", n_colors=len(MODEL_ORDER))
    MODEL_COLORS = dict(zip(MODEL_ORDER, _palette_raw))

    BASE_FONT_SIZE = 18
    DPI = 600
    YLABEL_PADDING = 15

    if feature_label_mapping is None:
        feature_label_mapping = {}

    sns.set_theme(style="whitegrid", context="talk")
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']
    plt.rcParams['figure.dpi'] = DPI

    # --- Strategy titles ---
    strategy_titles = {
        "strategy_one": "Prompting Strategy One",
        "strategy_two": "Prompting Strategy Two",
        "strategy_three": "Prompting Strategy Three",
        "strategy_four": "Prompting Strategy Four"
    }

    strategy_base_features = {
        "strategy_one": ["inf_diff_z", "bc_diff_z"],
        "strategy_two": ["inf_diff_z", "bc_diff_z", "cap_z"],
        "strategy_three": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z"],
        "strategy_four": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z", "preferred_option_num_z"]
    }

    # -----------------------------
    # Build global feature list (union across panels) in a stable order
    # -----------------------------
    ordered = []
    ordered_set = set()
    observed = set()

    for strategy_name in strategy_base_features.keys():
        df = all_results[strategy_name]["odds_ratios_df"]
        if df is not None and not df.empty and "feature" in df.columns:
            observed.update(df["feature"].dropna().unique().tolist())

        feats_imp = all_results[strategy_name].get("features_important_to_any_model", None)
        if feats_imp:
            for f in feats_imp:
                if f not in ordered_set:
                    ordered.append(f)
                    ordered_set.add(f)

    remaining = sorted([f for f in observed if f not in ordered_set])
    global_features = ordered + remaining

    if len(global_features) == 0:
        raise ValueError("No features found in any panel (all odds_ratios_df empty or missing 'feature').")

    # -----------------------------
    # Determine global x-range
    # -----------------------------
    _user_xmin = global_xmin is not None
    _user_xmax = global_xmax is not None

    if not _user_xmin or not _user_xmax:
        inferred_lows = []
        inferred_highs = []

        for strategy_name in strategy_base_features.keys():
            df = all_results[strategy_name]["odds_ratios_df"]
            if df is None or df.empty:
                continue
            for low_col in ["ci_lower", "ci_low", "lower", "lower_ci"]:
                if low_col in df.columns:
                    inferred_lows.extend(df[low_col].dropna().tolist())
                    break
            for high_col in ["ci_upper", "ci_high", "upper", "upper_ci"]:
                if high_col in df.columns:
                    inferred_highs.extend(df[high_col].dropna().tolist())
                    break

        # Fall back to point estimates only if CI columns are genuinely absent
        if len(inferred_lows) == 0 or len(inferred_highs) == 0:
            all_ors = []
            for strategy_name in strategy_base_features.keys():
                df = all_results[strategy_name]["odds_ratios_df"]
                if df is not None and not df.empty and "odds_ratio" in df.columns:
                    all_ors.extend(df["odds_ratio"].dropna().tolist())
            if len(inferred_lows) == 0:
                inferred_lows = all_ors.copy()
            if len(inferred_highs) == 0:
                inferred_highs = all_ors.copy()

        inferred_min = float(np.nanmin(inferred_lows)) if inferred_lows else 0.5
        inferred_max = float(np.nanmax(inferred_highs)) if inferred_highs else 2.0

        if not _user_xmin:
            global_xmin = inferred_min
        if not _user_xmax:
            global_xmax = inferred_max

    eps = 1e-6
    global_xmin = max(float(global_xmin), eps)
    global_xmax = max(float(global_xmax), global_xmin * 1.01)

    use_log = (global_xmax / global_xmin) > 10

    # Apply padding only when bounds were inferred; respect user-provided bounds exactly
    if _user_xmin and _user_xmax:
        global_xlim = (global_xmin, global_xmax)
    elif not use_log:
        global_xlim = (
            global_xmin - 0.15 * (global_xmax - global_xmin),
            global_xmax + 0.05 * (global_xmax - global_xmin)
        )
    else:
        global_xlim = (global_xmin / 1.4, global_xmax * 1.1)

    # -----------------------------
    # Figure / axes
    # -----------------------------
    n_cols = 4
    fig = plt.figure(figsize=(25, row_height))
    gs = fig.add_gridspec(1, n_cols, hspace=1, wspace=column_spacing)
    axes = [fig.add_subplot(gs[0, j]) for j in range(n_cols)]

    # Legend handles
    all_models_in_data = set()
    for strategy_name in strategy_base_features.keys():
        df = all_results[strategy_name]["odds_ratios_df"]
        if df is not None and not df.empty and "Model" in df.columns:
            all_models_in_data.update(df["Model"].unique())

    all_models_display = [MODEL_NAME_MAPPING.get(m, m) for m in all_models_in_data]
    available_legend_models = [m for m in MODEL_ORDER if m in all_models_display]
    legend_handles = [
        mpatches.Patch(facecolor=MODEL_COLORS[model], edgecolor='black', alpha=0.85, label=model)
        for model in available_legend_models
    ]

    palette = MODEL_COLORS

    # Constant y positions (global, shared across all panels)
    n_features_global = len(global_features)
    y_positions_global = np.arange(n_features_global) * 1.2

    def _apply_axes_styling(ax, idx):
        """Apply shared x-axis, y-axis, and grid styling to an axis."""
        ax.set_xlabel('Odds Ratio (95% CI)', fontweight='bold', fontsize=BASE_FONT_SIZE * 1.0)

        if idx == 0:
            ax.set_ylabel('Feature', fontweight='bold', fontsize=BASE_FONT_SIZE * 1.1, labelpad=YLABEL_PADDING)
        else:
            ax.set_ylabel('')

        ax.set_yticks(y_positions_global)
        feature_labels = [feature_label_mapping.get(f, f) for f in global_features]
        ax.set_yticklabels(feature_labels, fontsize=BASE_FONT_SIZE * 0.9)

        # Lock y-axis limits so all panels share the exact same viewport
        y_margin = 0.6
        ax.set_ylim(
            y_positions_global[0] - y_margin,
            y_positions_global[-1] + y_margin
        )

        ax.axvline(x=1, color='black', linestyle='--', linewidth=2, alpha=0.7, zorder=1)

        ax.set_xscale('log' if use_log else 'linear')
        ax.set_xlim(*global_xlim)

        if use_log:
            ax.xaxis.set_major_locator(mticker.LogLocator(base=10.0, numticks=6))
            ax.xaxis.set_minor_locator(mticker.LogLocator(base=10.0, subs=np.arange(2, 10) * 0.1, numticks=12))
            ax.xaxis.set_major_formatter(mticker.LogFormatterSciNotation(labelOnlyBase=True))
            ax.tick_params(axis='x', which='minor', length=3, width=0.8)
        else:
            tick_vals = np.linspace(global_xlim[0], global_xlim[1], 6)
            ax.set_xticks(tick_vals)
            ax.xaxis.set_major_formatter(mticker.FormatStrFormatter('%.2g'))

        ax.tick_params(axis='x', which='major', labelsize=BASE_FONT_SIZE * 0.85, length=5, width=1.2)
        ax.grid(True, axis='x', alpha=0.25, linestyle='--')
        ax.grid(False, axis='y')
        ax.set_axisbelow(True)
        ax.tick_params(axis='both', labelsize=BASE_FONT_SIZE * 0.9)

    for idx, (strategy_name, base_features) in enumerate(strategy_base_features.items()):
        ax = axes[idx]
        odds_ratios_df = all_results[strategy_name]["odds_ratios_df"]

        ax.set_title(
            strategy_titles[strategy_name],
            fontweight='bold',
            pad=50,
            fontsize=BASE_FONT_SIZE * 1.3
        )

        if odds_ratios_df is None or odds_ratios_df.empty:
            _apply_axes_styling(ax, idx)
            continue

        odds_ratios_df = odds_ratios_df.copy()
        odds_ratios_df['Model_Display'] = odds_ratios_df['Model'].map(MODEL_NAME_MAPPING)
        odds_ratios_df['Model_Display'] = odds_ratios_df['Model_Display'].fillna(odds_ratios_df['Model'])

        # Pivot
        pivot_or = odds_ratios_df.pivot(index='feature', columns='Model_Display', values='odds_ratio')
        pivot_ci_low = odds_ratios_df.pivot(index='feature', columns='Model_Display', values='ci_lower')
        pivot_ci_high = odds_ratios_df.pivot(index='feature', columns='Model_Display', values='ci_upper')

        # Reorder columns (models)
        available_models = [m for m in MODEL_ORDER if m in pivot_or.columns]
        pivot_or = pivot_or[available_models]
        pivot_ci_low = pivot_ci_low[available_models]
        pivot_ci_high = pivot_ci_high[available_models]

        # Reindex rows to global_features — missing features become NaN (blank rows)
        pivot_or = pivot_or.reindex(global_features)
        pivot_ci_low = pivot_ci_low.reindex(global_features)
        pivot_ci_high = pivot_ci_high.reindex(global_features)

        n_models = len(pivot_or.columns)
        spacing = 0.18

        for i, model in enumerate(pivot_or.columns):
            offset = (i - n_models / 2 + 0.5) * spacing
            color = palette.get(model, 'gray')

            or_vals = pivot_or[model].values
            ci_low_vals = pivot_ci_low[model].values
            ci_high_vals = pivot_ci_high[model].values

            for j in range(n_features_global):
                if pd.notna(or_vals[j]):
                    or_val = float(or_vals[j])
                    ci_low = float(ci_low_vals[j]) if pd.notna(ci_low_vals[j]) else or_val
                    ci_high = float(ci_high_vals[j]) if pd.notna(ci_high_vals[j]) else or_val

                    left_err = max(0.0, or_val - ci_low)
                    right_err = max(0.0, ci_high - or_val)

                    ax.errorbar(
                        or_val, y_positions_global[j] + offset,
                        xerr=[[left_err], [right_err]],
                        fmt='o',
                        color=color,
                        ecolor=color,
                        markersize=dot_size,
                        capsize=5,
                        alpha=0.85,
                        linewidth=error_bar_width,
                        markeredgecolor='black',
                        markeredgewidth=0.8
                    )

        _apply_axes_styling(ax, idx)

    fig.legend(
        handles=legend_handles,
        title="Model",
        loc="upper center",
        bbox_to_anchor=(0.5, 1.28),
        ncol=max(1, len(available_legend_models)),
        fontsize=BASE_FONT_SIZE * 0.95,
        title_fontsize=BASE_FONT_SIZE * 1.05,
        frameon=True,
        edgecolor="black",
        fancybox=False
    )

    sns.despine()
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.show()
```

```python
def visualize_elasticnet_forest_plots(
    all_results,
    # Visualization params
    row_height=6.0,
    column_spacing=0.5,
    dot_size=10,
    error_bar_width=2.5,
    feature_label_mapping=None,
    global_xmin=None,
    global_xmax=None,
    wrap_y_labels_at=None,  # optional wrap length (None = no wrap)
    y_order=None             # list of feature names (top -> bottom). If None use automatic ordering.
):
    """
    Visualize Elastic-Net odds ratios as forest plots in a single row.

    Updates:
      - If global_xmin is not provided, the x-axis lower bound is forced to start at 1e-9.
      - All other behavior remains compatible with the original function.
    """
    import pandas as pd
    import numpy as np
    import matplotlib.pyplot as plt
    import matplotlib.patches as mpatches
    import matplotlib.ticker as mticker
    import seaborn as sns
    import textwrap

    # --- Styling Configuration ---
    MODEL_NAME_MAPPING = {
        'x-ai/grok-4-fast': 'Grok',
        'openai/gpt-5-nano': 'ChatGPT',
        'google/gemini-2.5-flash': 'Gemini',
        'deepseek/deepseek-chat-v3.1': 'DeepSeek'
    }
    MODEL_ORDER = ['ChatGPT', 'DeepSeek', 'Gemini', 'Grok']

    _palette_raw = sns.color_palette("tab10", n_colors=len(MODEL_ORDER))
    MODEL_COLORS = dict(zip(MODEL_ORDER, _palette_raw))

    BASE_FONT_SIZE = 18
    DPI = 600
    YLABEL_PADDING = 15

    if feature_label_mapping is None:
        feature_label_mapping = {}

    sns.set_theme(style="whitegrid", context="talk")
    plt.rcParams['font.family'] = 'sans-serif'
    plt.rcParams['font.sans-serif'] = ['Arial', 'DejaVu Sans']
    plt.rcParams['figure.dpi'] = DPI

    # --- Strategy titles & base features ---
    strategy_titles = {
        "strategy_one": "Prompting Strategy One",
        "strategy_two": "Prompting Strategy Two",
        "strategy_three": "Prompting Strategy Three",
        "strategy_four": "Prompting Strategy Four"
    }

    strategy_base_features = {
        "strategy_one": ["inf_diff_z", "bc_diff_z"],
        "strategy_two": ["inf_diff_z", "bc_diff_z", "cap_z"],
        "strategy_three": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z"],
        "strategy_four": ["inf_diff_z", "bc_diff_z", "cap_z", "income_loss_diff_z", "preferred_option_num_z"]
    }

    # -----------------------------
    # Build global feature list (union across panels) in a stable order
    # -----------------------------
    ordered = []
    ordered_set = set()
    observed = set()

    for strategy_name in strategy_base_features.keys():
        df = all_results[strategy_name].get("odds_ratios_df", None)
        if df is not None and not df.empty and "feature" in df.columns:
            observed.update(df["feature"].astype(str).str.strip().unique().tolist())

        feats_imp = all_results[strategy_name].get("features_important_to_any_model", None)
        if feats_imp:
            for f in feats_imp:
                fnorm = str(f).strip()
                if fnorm not in ordered_set:
                    ordered.append(fnorm)
                    ordered_set.add(fnorm)

    remaining = sorted([f for f in observed if f not in ordered_set])
    auto_global_features = ordered + remaining

    if len(auto_global_features) == 0:
        raise ValueError("No features found in any panel (all odds_ratios_df empty or missing 'feature').")

    # -----------------------------
    # Apply optional user-specified y order
    # -----------------------------
    if y_order is not None:
        user_order_norm = [str(x).strip() for x in list(y_order)]
        seen = set()
        user_order_unique = []
        for u in user_order_norm:
            if u not in seen:
                user_order_unique.append(u)
                seen.add(u)
        appended = [f for f in auto_global_features if f not in seen]
        global_features = user_order_unique + appended
    else:
        gf = list(auto_global_features)
        gf = list(reversed(gf))
        gf_with_idx = [(f, idx) for idx, f in enumerate(gf)]

        def degree_key(item):
            feat, orig_idx = item
            degree = str(feat).count("_x_")
            return (degree, orig_idx)

        gf_sorted = sorted(gf_with_idx, key=degree_key)
        global_features = [f for f, _ in gf_sorted]

    # -----------------------------
    # Determine global x-range
    # -----------------------------
    _user_xmin = global_xmin is not None
    _user_xmax = global_xmax is not None

    if not _user_xmin or not _user_xmax:
        inferred_lows = []
        inferred_highs = []

        for strategy_name in strategy_base_features.keys():
            df = all_results[strategy_name].get("odds_ratios_df", None)
            if df is None or df.empty:
                continue

            for low_col in ["ci_lower", "ci_low", "lower", "lower_ci"]:
                if low_col in df.columns:
                    inferred_lows.extend(pd.to_numeric(df[low_col], errors='coerce').dropna().tolist())
                    break

            for high_col in ["ci_upper", "ci_high", "upper", "upper_ci"]:
                if high_col in df.columns:
                    inferred_highs.extend(pd.to_numeric(df[high_col], errors='coerce').dropna().tolist())
                    break

        # fallback to odds if CI absent
        if len(inferred_lows) == 0 or len(inferred_highs) == 0:
            all_ors = []
            for strategy_name in strategy_base_features.keys():
                df = all_results[strategy_name].get("odds_ratios_df", None)
                if df is not None and not df.empty and "odds_ratio" in df.columns:
                    all_ors.extend(pd.to_numeric(df["odds_ratio"], errors='coerce').dropna().tolist())
            if len(inferred_lows) == 0:
                inferred_lows = all_ors.copy()
            if len(inferred_highs) == 0:
                inferred_highs = all_ors.copy()

        inferred_min = float(np.nanmin(inferred_lows)) if inferred_lows else 0.5
        inferred_max = float(np.nanmax(inferred_highs)) if inferred_highs else 2.0

        if not _user_xmin:
            global_xmin = min(inferred_min, 1e-9)
        if not _user_xmax:
            global_xmax = inferred_max

    eps = 1e-12
    global_xmin = max(float(global_xmin), eps)
    global_xmax = max(float(global_xmax), global_xmin * 1.01)

    use_log = (global_xmax / global_xmin) > 10

    if _user_xmin and _user_xmax:
        global_xlim = (global_xmin, global_xmax)
    elif not use_log:
        span = (global_xmax - global_xmin) if global_xmax > global_xmin else max(abs(global_xmin), 1.0)
        global_xlim = (
            global_xmin - 0.15 * span,
            global_xmax + 0.05 * span
        )
    else:
        global_xlim = (global_xmin, global_xmax * 1.1)

    # -----------------------------
    # Figure / axes
    # -----------------------------
    n_cols = 4
    fig = plt.figure(figsize=(25, row_height))
    gs = fig.add_gridspec(1, n_cols, hspace=1, wspace=column_spacing)
    axes = [fig.add_subplot(gs[0, j]) for j in range(n_cols)]

    # Legend handles
    all_models_in_data = set()
    for strategy_name in strategy_base_features.keys():
        df = all_results[strategy_name].get("odds_ratios_df", None)
        if df is not None and not df.empty and "Model" in df.columns:
            all_models_in_data.update(df["Model"].unique())

    all_models_display = [MODEL_NAME_MAPPING.get(m, m) for m in all_models_in_data]
    available_legend_models = [m for m in MODEL_ORDER if m in all_models_display]
    legend_handles = [
        mpatches.Patch(facecolor=MODEL_COLORS[model], edgecolor='black', alpha=0.85, label=model)
        for model in available_legend_models
    ]

    palette = MODEL_COLORS

    # Constant y positions shared across panels
    n_features_global = len(global_features)
    y_positions_global = np.arange(n_features_global) * 1.2

    def wrap_label(text):
        lab = str(text)
        if wrap_y_labels_at is None:
            return lab
        return "\n".join(textwrap.wrap(lab, width=int(wrap_y_labels_at)))

    def _apply_axes_styling(ax, idx, present_features):
        ax.set_xlabel('Odds Ratio (95% CI)', fontweight='bold', fontsize=BASE_FONT_SIZE * 1.0)

        if idx == 0:
            ax.set_ylabel('Feature', fontweight='bold', fontsize=BASE_FONT_SIZE * 1.1, labelpad=YLABEL_PADDING)
        else:
            ax.set_ylabel('')

        ax.set_yticks(y_positions_global)

        labels = []
        pf_norm = set([str(f).strip() for f in present_features])
        for f in global_features:
            f_norm = str(f).strip()
            if f_norm in pf_norm:
                disp = feature_label_mapping.get(f, feature_label_mapping.get(f_norm, f_norm))
                labels.append(wrap_label(disp))
            else:
                labels.append("")

        ax.set_yticklabels(labels, fontsize=BASE_FONT_SIZE * 0.9)

        y_margin = 0.6
        ax.set_ylim(y_positions_global[0] - y_margin, y_positions_global[-1] + y_margin)

        ax.axvline(x=1, color='black', linestyle='--', linewidth=2, alpha=0.7, zorder=1)

        ax.set_xscale('log' if use_log else 'linear')
        ax.set_xlim(*global_xlim)

        if use_log:
            ax.xaxis.set_major_locator(mticker.LogLocator(base=10.0, numticks=10))
            ax.xaxis.set_minor_locator(mticker.LogLocator(base=10.0, subs=np.arange(2, 10) * 0.1, numticks=12))
            ax.xaxis.set_major_formatter(mticker.LogFormatterSciNotation(labelOnlyBase=True))
            ax.tick_params(axis='x', which='minor', length=3, width=0.8)
        else:
            tick_vals = np.linspace(global_xlim[0], global_xlim[1], 6)
            ax.set_xticks(tick_vals)
            ax.xaxis.set_major_formatter(mticker.FormatStrFormatter('%.2g'))

        ax.tick_params(axis='x', which='major', labelsize=BASE_FONT_SIZE * 0.85, length=5, width=1.2)
        ax.grid(True, axis='x', alpha=0.25, linestyle='--')
        ax.grid(False, axis='y')
        ax.set_axisbelow(True)
        ax.tick_params(axis='both', labelsize=BASE_FONT_SIZE * 0.9)

    # -----------------------------
    # Plot panels
    # -----------------------------
    for idx, (strategy_name, base_features) in enumerate(strategy_base_features.items()):
        ax = axes[idx]
        odds_ratios_df = all_results[strategy_name].get("odds_ratios_df", None)

        ax.set_title(strategy_titles[strategy_name], fontweight='bold', pad=50, fontsize=BASE_FONT_SIZE * 1.3)

        if odds_ratios_df is None or odds_ratios_df.empty:
            _apply_axes_styling(ax, idx, present_features=set())
            continue

        odds_ratios_df = odds_ratios_df.copy()
        odds_ratios_df['feature'] = odds_ratios_df['feature'].astype(str).str.strip()
        odds_ratios_df['Model_Display'] = odds_ratios_df['Model'].map(MODEL_NAME_MAPPING).fillna(odds_ratios_df['Model'])

        pivot_or = odds_ratios_df.pivot(index='feature', columns='Model_Display', values='odds_ratio')
        pivot_ci_low = odds_ratios_df.pivot(index='feature', columns='Model_Display', values='ci_lower')
        pivot_ci_high = odds_ratios_df.pivot(index='feature', columns='Model_Display', values='ci_upper')

        available_models = [m for m in MODEL_ORDER if m in pivot_or.columns]
        pivot_or = pivot_or.reindex(columns=available_models)
        pivot_ci_low = pivot_ci_low.reindex(columns=available_models)
        pivot_ci_high = pivot_ci_high.reindex(columns=available_models)

        gf_for_reindex = [str(f).strip() for f in global_features]
        pivot_or = pivot_or.reindex(gf_for_reindex)
        pivot_ci_low = pivot_ci_low.reindex(gf_for_reindex)
        pivot_ci_high = pivot_ci_high.reindex(gf_for_reindex)

        n_models = len(pivot_or.columns)
        spacing = 0.18

        present_features = set(pivot_or.dropna(how='all').index.tolist())

        for i, model in enumerate(pivot_or.columns):
            offset = (i - n_models / 2 + 0.5) * spacing
            color = palette.get(model, 'gray')

            or_vals = pivot_or[model].values
            ci_low_vals = pivot_ci_low[model].values
            ci_high_vals = pivot_ci_high[model].values

            for j in range(n_features_global):
                if pd.notna(or_vals[j]):
                    or_val = float(or_vals[j])
                    ci_low = float(ci_low_vals[j]) if not pd.isna(ci_low_vals[j]) else or_val
                    ci_high = float(ci_high_vals[j]) if not pd.isna(ci_high_vals[j]) else or_val

                    left_err = max(0.0, or_val - ci_low)
                    right_err = max(0.0, ci_high - or_val)

                    ax.errorbar(
                        or_val,
                        y_positions_global[j] + offset,
                        xerr=[[left_err], [right_err]],
                        fmt='o',
                        color=color,
                        ecolor=color,
                        markersize=dot_size,
                        capsize=5,
                        alpha=0.85,
                        linewidth=error_bar_width,
                        markeredgecolor='black',
                        markeredgewidth=0.8
                    )

        _apply_axes_styling(ax, idx, present_features=present_features)

    # Legend & finishing
    fig.legend(
        handles=legend_handles,
        loc="upper center",
        bbox_to_anchor=(0.5, 1.18),
        ncol=max(1, len(available_legend_models)),
        fontsize=BASE_FONT_SIZE * 0.95,
        title_fontsize=BASE_FONT_SIZE * 1.05,
        frameon=True,
        edgecolor="black",
        fancybox=False
    )

    sns.despine()
    plt.tight_layout()
    plt.subplots_adjust(top=0.92)
    plt.show()
```
